```python
# looping
input_dict = {'Name':'Aman','Age':20,"Skills":'Python'}
```

```python
# Keys Printing
for x in input_dict:
    print(x)
```

```python
# Values Printing
for x in input_dict:
    print(input_dict[x])
```

```python
# Getting Both
for x,y in input_dict.items():
    print(x,y)
```

```python
# Copy Dictionary but its risky as it will make changes in the original dict also
copy_dict = input_dict
print(copy_dict)
```

```python
copy_dict['Name'] = 'Kiran'
```

```python
input_dict
```

```python
# Using copy function
input_dict = {'Name':'Aman','Age':20,"Skills":'Python'}
copy_dict = input_dict.copy()
copy_dict
```

```python
copy_dict['Name'] = 'Suman'
```
