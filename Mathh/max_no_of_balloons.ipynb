```python
text = "nlaebolko"
freq_text = {}
freq_balloon = {}

for ch in text:
    freq_text[ch] = freq_text.get(ch, 0) + 1
print(freq_text)

for ch in "balloon":    
    freq_balloon[ch] = freq_balloon.get(ch, 0) + 1
print(freq_balloon)


ans = float('inf')

for ch,cnt in freq_balloon.items():
    ans = min(ans, freq_text.get(ch, 0) // cnt)
print(ans)
```
