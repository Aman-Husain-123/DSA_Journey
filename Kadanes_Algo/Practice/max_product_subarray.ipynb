```python
nums = [2,3,-2,4]
min_ending_here = max_ending_here = max_so_far = nums[0]

for i in range(1, len(nums)):
    v1 = nums[i]
    v2 = max_ending_here * nums[i]
    v3 = min_ending_here * nums[i]
    max_ending_here = max(v1, max(v2, v3))
    min_ending_here = min(v1, min(v2, v3))
    res = max(max_so_far, max_ending_here)
print(res)
```
