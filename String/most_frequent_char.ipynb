```python
s = "testsample"
mapp = {}
res = []
maxx = 0
for i in range(len(s)):
    mapp[s[i]] = mapp.get(s[i], 0) + 1

for key, value in mapp.items():
    if value > maxx:
        maxx = value
        res = [key]
    elif value == maxx:
        res.append(key)
print(min(res))
```

```python
# Optimal Approach
s = "testsample"

stack = []

for ch in s:
    if stack and stack[-1][0] == ch:
        char, freq = stack.pop()
        stack.append((char, freq + 1))
    else:
        stack.append((ch, 1))

print(stack)
```

```python
def main():
    n = 3
    data = [
        "A","B","C","10",
        "A","B","C","20",
        "A","B","C","30"
    ]
    
    idx = 0
    seen = set()
    duplicates = []
    
    for _ in range(n):
        a = data[idx]; 
        b = data[idx+1]; 
        c = data[idx+2]; 
        time = data[idx+3]
        idx += 4
        
        key = (a, b, c)
        
        if key in seen:
            duplicates.append((a, b, c, time))
        else:
            seen.add(key)
    
    for record in duplicates:
        print(*record)

main()
```

```python
# TODO: Fraud Checking By Transaction By Time
from collections import defaultdict

def detect_fraud(transactions):
    # transactions = [(sender, receiver, amount, timestamp)]

    pair_map = defaultdict(list)
    fraud_transactions = []

    # Step 1: Group by (sender, receiver)
    for sender,receiver,amount,time in transactions:
        pair_map[(sender,receiver)].append((time,amount))
    
    # Step 2: Process each pair
    for (sender,receiver),values in pair_map.items():
        # Sort by time
        values.sort()

        # Step 3: Compare consecutive transactions
        for i in range(1, len(values)):
            if values[i][0] - values[i-1][0] < 30:
                fraud_transactions.append(
                    (sender, receiver, values[i][1], values[i][0])
                )
        print(pair_map)
    
    return fraud_transactions


# Example usage
transactions = [
    ("A", "B", 100, 10),
    ("A", "B", 200, 20),   # fraud (20 - 10 < 30)
    ("A", "C", 300, 100),
    ("A", "B", 400, 60),   # not fraud (60 - 20 > 30)
]

result = detect_fraud(transactions)

for txn in result:
    print(*txn)
```

```python
# TODO: Toggling Bits
def toggle_bits(n):
    # count number of bits
    k = len(bin(n)) - 2   # remove '0b'
    print(k)
    
    # create mask with all bits = 1
    mask = (1 << k) - 1
    print(mask)
    
    # toggle bits
    return mask - n
    

# Input
n = int(input().strip())

# Output
print(toggle_bits(n))
```
