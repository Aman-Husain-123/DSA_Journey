```python
s = "testsample"
mapp = {}
res = []
maxx = 0
for i in range(len(s)):
    mapp[s[i]] = mapp.get(s[i], 0) + 1

for key, value in mapp.items():
    if value > maxx:
        maxx = value
        res = [key]
    elif value == maxx:
        res.append(key)
print(min(res))
```

```python
# Optimal Approach
s = "testsample"

stack = []

for ch in s:
    if stack and stack[-1][0] == ch:
        char, freq = stack.pop()
        stack.append((char, freq + 1))
    else:
        stack.append((ch, 1))

print(stack)
```
