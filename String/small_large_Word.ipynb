```python
s = "Hi from Gfg"
res = []
words = s.split()

min_word = words[0]
max_word = words[0]

for word in words:
    if len(word) < len(min_word):
        min_word = word
            
    if len(word) >= len(max_word):
        max_word = word
        
res.append(max_word)
res.append(min_word)
print(res)
```
