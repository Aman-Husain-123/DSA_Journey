```python
arr = [1, 2, 1, 3, 4, 2, 3]
k = 4

freq = {}
result = []

for i in range(k):
    freq[arr[i]] = freq.get(arr[i], 0) + 1

result.append(len(freq))

for i in range(k, len(arr)):
    left = arr[i-k]
    freq[left] -= 1

    if freq[left] == 0:
        del freq[left]
    
    freq[arr[i]] = freq.get(arr[i], 0) + 1
    result.append(len(freq))
print(result)
```
