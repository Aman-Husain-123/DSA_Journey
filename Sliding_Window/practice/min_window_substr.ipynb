```python
s = "ADOBECODEBANC"
t = "ABC"

# Edge Case: If either string is empty, return an empty string
if not s or not t:
    print("")

# Step 1: Create a frequency map for characters in t
# Example for t = "ABC", the frequency map would be {'A': 1, 'B': 1, 'C': 1}
need = {}
for char in t:
    need[char] = need.get(char,0) + 1

# This will keep track of how many characters we have in the current window
window = {}

# Left pointer of the sliding window
left = 0

# Total char we still need to match
# (Counts duplicates also)
required = len(t)

#  # Variables to track the minimum window
min_length = float('inf')
start = 0

# Step 2: Expand the window using right pointer
for right in range(len(s)):
    # This is the character we are adding to the window
    char = s[right]

    # If current char is needed
    if char in need:
        # Add it to the window count
        window[char] = window.get(char, 0) + 1

        # if this char count is within required, decrease the required count
        if window[char] <= need[char]:
            required -= 1
    
    # Step 3: Try to contract the window from the left if we have all required chars
    while required == 0:
        # Update minimum window if the current one is smaller
        if right-left + 1 < min_length:
            min_length = right-left + 1
            # This is the starting index of the minimum window
            start = left
        
        # Remove the left char and shrink window
        left_char = s[left]

        # If the left char is needed, we need to update the window count and possibly increase required
        if left_char in need:
            window[left_char] -= 1

            # If after removing, the window no longer satisfies need,
            # increase required (window becomes invalid)
            if window[left_char] < need[left_char]:
                required += 1

        
        # Move left pointer to shrink the window
        left += 1


# Step 4: Return the minimum window substring
if min_length == float('inf'):
    print("")
else:
    print(s[start:start+min_length])
```
