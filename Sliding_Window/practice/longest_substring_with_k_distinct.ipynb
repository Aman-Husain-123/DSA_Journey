```python
s = "aabacbebebe"
k = 3

left = 0
frequency = {}
max_length = 0

for right in range(len(s)):
    # Add the current character to the frequency dictionary
    frequency[s[right]] = frequency.get(s[right], 0) + 1

    # If more than k unique characters, shrink the window from the left
    while len(frequency) > k:
        frequency[s[left]] -= 1
        if frequency[s[left]] == 0:
            del frequency[s[left]]
        left += 1
    
    # If exactly k unique characters, update the maximum length
    if len(frequency) == k:
        max_length = max(max_length,right-left+1)
print(max_length)
```
