```python
# When window is not fixed then keep high and low at the same position
# TC: O(n)
# SC: O(1)
arr = [1,1,1,1,1,1,1,1]
n = len(arr)
target = 11
low = 0
high = 0
sum = 0

res = 0

while high < n:
    sum += arr[high]

    while sum >= target:
        lenn = high - low + 1
        res = min(res, lenn)
        sum -= arr[low]
        low += 1
    high += 1

print("Current minimum length:", res)
```
