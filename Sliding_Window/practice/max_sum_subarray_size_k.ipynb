```python
arr = [100,200,300,400,500]
k = 2
low = 0
high = 1
sum = 0
result = 0

for i in range(low,high+1):
    sum += arr[i]

    while high < len(arr):
        result = max(result, sum)
        high += 1
        low += 1
        
        if high == len(arr):
            break

        sum = sum - arr[low-1] + arr[high]
print(result)
```
