```python
arr = [1,1,2,3,8,4,4]
target = 5
left, right = 0, len(arr) - 1
nums_with_index = list(enumerate(arr))
res = []

while left <= right:
    total = nums_with_index[left][1] + nums_with_index[right][1]
    if total == target:
        res.append(nums_with_index[left][0])
        res.append(nums_with_index[right][0])
    elif total < target:
        left += 1
    else:
        right -= 1
# print(nums_with_index)
```
