```python
arr = [100, 200, 300, 400]
k = 2

low = 0
high = 1
result = 0
summ = 0

for i in range(low,high + 1):
    summ += arr[i]

    result = summ

    while high < len(arr)-1:
        high += 1
        low += 1

        summ = summ - arr[low - 1] + arr[high]
        result = max(result, summ)
print(result)

```
