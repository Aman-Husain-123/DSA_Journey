```python
nums = [2,7,11,15]
target = 9
```

```python
nums_with_index = list(enumerate(nums))
nums_with_index.sort(key=lambda x:x[1])
```

```python
left,right = 0, len(nums)-1

while left < right:
    total = nums_with_index[left][1] + nums_with_index[right][1]

    if total == target:
        print([nums_with_index[left][0], nums_with_index[right][0]])
        break
    elif total < target:
        left += 1
    else:
        right -= 1
        
```
