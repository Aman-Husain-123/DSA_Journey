```python
nums = [-1,0,1,2,-1,-4]
nums.sort()
res = []

for i in range(len(nums)-1):
    for j in range(i+1, len(nums)-1):
        for k in range(j+1, len(nums)-1):
            if nums[i] + nums[j] + nums[k] == 0:
                res.append([nums[i], nums[j], nums[k]])
                
print(res)
```
