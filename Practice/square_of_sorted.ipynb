```python
nums = [-4,-1,0,3,10]

left, right = 0, len(nums) - 1
squared = [0] * len(nums)

for i in range(len(nums)-1,-1,-1):
    if abs(nums[left]) > abs(nums[right]):
        squared[i] = nums[left ]**2
        left += 1
    else:
        squared[i] = nums[right] ** 2
        right -= 1
print(squared)
```
