```python
# TODO: TC:- O(n^3)
# SC:- O(n)
arr = [-1,0,1,2,-1,-4]
arr.sort()
res = []
for i in range(len(arr)):
    for j in range(i+1, len(arr)):
        for k in range(j+1, len(arr)):
            if arr[i] + arr[j] + arr[k] == 0:
                res.append([arr[i], arr[j], arr[k]])
print(res)
```

```python
# TODO: 2 Pointer Approach
# TC:- O(n^2)
# SC:- O(n)
arr = [-1,0,1,2,-1,-4]
arr.sort()
res = []
for i in range(len(arr)):
    l = i + 1
    r = len(arr) - 1
    while l < r:
        if arr[i] + arr[l] + arr[r] == 0:
            res.append([arr[i], arr[l], arr[r]])
            l += 1
            r -= 1
        elif arr[i] + arr[l] + arr[r] < 0:
            l += 1
        else:
            r -= 1
print(res)
```

```python
arr = [-1,0,1,2,-1,-4]
arr.sort()

left = 0
right = len(arr) - 1
res = []
target = 0

while left < right:
    summ = arr[left] + arr[right]

    if summ == target:
        res.append([arr[left], arr[right]])
        left += 1
        right -= 1

        # Skip duplicates
        while left < right and arr[left] == arr[left - 1]:
            left += 1
        while left < right and arr[right] == arr[right + 1]:
            right -= 1
    elif summ < target:
        left += 1
    else:
        right -= 1
print(res)
```
