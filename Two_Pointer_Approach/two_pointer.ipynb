```python
# Brute Approach
arr = [1,2,1,3,4,2]
target = 3
for i in range(len(arr)):
    for j in range(i+1, len(arr)):
        if arr[i] + arr[j] == target:
            print(arr[i],arr[j])
```

```python
# Hasmap Approach
# TODO: TC :- O(n) brcause we are traversing the array only once 
# TODO: and SC :- O(n)
arr = [1,2,1,3,4,2]
target = 3
hashmap = {}
for i in range(len(arr)):
    # Calculate complement (e,g:- if target is 3 and arr[i] is 1 then complement will be 2)
    complement = target - arr[i]

    # Check if complement exists in hashmap
    if complement in hashmap:
        print([hashmap[complement],i])
        break

        # Store the current element's index in the hashmap
    hashmap[arr[i]] = i

print(hashmap)
```

```python
# TODO: Two Pointer Approach
arr = [1,2,1,3,4,2]
target = 3
# Step 1: Sort the array
arr.sort()  
# Step 2: Initialize two pointers
left = 0
res = []
right = len(arr) - 1
# Step 3: Iterate until the pointersr meet
while left < right:
    current_sum = arr[left] + arr[right]
    if current_sum == target:
        res.append([left, right])  # Store the pair in the result list
        break
        left += 1  # Move left pointer to the right
        right -= 1  # Move right pointer to the left
    elif current_sum < target:
        left += 1  # Move left pointer to the right to increase sum
    else:
        right -= 1  # Move right pointer to the left to decrease sum
res
```
