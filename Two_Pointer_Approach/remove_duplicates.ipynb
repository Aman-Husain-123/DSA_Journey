```python
# TODO:Two Pointer Approach
        # TC: O(n)
        # SC :- O(1)
# Officer And CM Example Is Used for house allotment
nums = [1,1,2]
i = 0
j = 1
no_of_unqiue = 1

while j < len(nums):
    if nums[j] == nums[j-1]:
        j += 1
    else:
        nums[i+1] = nums[j]
        i += 1
        j += 1
        no_of_unqiue += 1
print(len(nums[:no_of_unqiue])) 
```
