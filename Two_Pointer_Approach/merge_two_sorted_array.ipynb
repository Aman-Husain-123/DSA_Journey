```python
arr2 = [1,3,5]
arr1 = [2,4,6]
res = [0] * (len(arr1) + len(arr2))
idx = 0

i,j = 0,0
while i < len(arr1) and j < len(arr2):
    if arr1[i] < arr2[j]:
        res[idx] = arr1[i]
        idx += 1
        i += 1
    else:
        res[idx] = arr2[j]
        idx += 1
        j += 1

while j < len(arr2):
    res[idx] = arr2[j]
    idx += 1
    j += 1

while i < len(arr1):
    res[idx] = arr1[i]
    idx += 1
    i += 1
print(res)
```
