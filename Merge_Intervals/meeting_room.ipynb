```python
start = [2, 9, 6]
end = [4, 12, 10]

start.sort()
n = len(start)
room = 0
res = 0

end.sort()
m = len(end)

i = j = 0

while i < n and j < m:
    if start[i] < end[j]:
        room += 1
        res = max(res, room)
        # print(f"Start: {start[i]}")
        i += 1
    else:
        room -= 1
        # print(f"End: {end[j]}")
        j += 1
print(res)
```
