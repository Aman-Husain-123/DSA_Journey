```python
arr = [[1, 3], [5, 7], [2, 4], [6, 8]]
arr.sort()

# Initialize with the first interval
start1 = arr[0][0]
end1 = arr[0][1]

# Start checking from the second interval
for i in range(1, len(arr)):
    start2 = arr[i][0]
    end2 = arr[i][1]

    # If current interval starts before or at the end of previous,
    # an intersection exists
    if start2 <= end1:
        print(True)
        break
    else:
        print(False)
    # Update the previous interval range
    start1 = start2
    end1 = end2
```
