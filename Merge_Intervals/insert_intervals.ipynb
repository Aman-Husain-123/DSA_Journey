```python
from typing import List


def insert(arr, newInterval) -> List[List[int]]:
        
        res = []
        n = len(arr)
        i = 0

        # 1. Add all intervals before newInterval
        while i < n and arr[i][1] < newInterval[0]:
            res.append(arr[i])
            i += 1

        # 2. Initialize merge window with newInterval
        start1 = newInterval[0]
        end1 = newInterval[1]
        print(arr)

        # 3. Merge overlapping intervals
        while i < n and arr[i][0] <= end1:
            start1 = min(start1, arr[i][0])
            end1 = max(end1, arr[i][1])
            i += 1

        res.append([start1, end1])

        # 4. Add remaining intervals
        while i < n:
            res.append(arr[i])
            i += 1

        return res
```
