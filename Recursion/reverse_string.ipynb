```python
strr = "aman"
rev_str = ""

for i in range(len(strr)-1,-1,-1):
    rev_str += strr[i]  
print(rev_str)
```

### `Into Recursion`

```python
def reverse_string_recursive(s):
    # Base Case
    if len(s) == 0:
        return s
    else:
        print("Current string:", s)
        return reverse_string_recursive(s[1:]) + s[0]
```

```python
print(reverse_string_recursive("aman"))
```
