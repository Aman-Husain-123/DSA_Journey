```python
# For a given range L to U, print all numbers from L to U using recursion.
def printNum(Lrange,Urange):
    # Base Case
    if Lrange > Urange:
        return
    
    print(Lrange)
    printNum(Lrange + 1, Urange) # Tail recursion because function call is the last statement
```

```python
print(printNum(1,5))
```

```python
# For reverse order
def printNumReverse(Lrange,Urange):
    # Base Case
    if Lrange > Urange:
        return
    
    printNumReverse(Lrange + 1,Urange)   # Head recursion because function call is the last aboves
    print(Lrange)
```

```python
printNumReverse(1,5)
```
