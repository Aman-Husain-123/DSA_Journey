```python
n = 4
k = 2

for i in range(1, k + 1):
    n = n * (k - i + 1) // i
print(n)
```

```python
def print_increasing_sequences(n, k, start=1, curr=[]):
    if k == 0:
        print(curr)
        return
    for i in range(start, n + 1):
        print_increasing_sequences(n, k - 1, i + 1, curr + [i])

# Example usage:
n = 4
k = 2
print_increasing_sequences(n, k)
```
