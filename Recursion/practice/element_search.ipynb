```python
def element_search(arr,index,target):
    if index >= len(arr):
        return -1
    if arr[index] == target:
        return index
    
    return element_search(arr, index + 1, target)
```

```python
arr = [10,20,30,40,50]
print(f"Array:",arr)
print("Indexes:",end = " ")

for j in range(len(arr)):
    print(j,end = " ")

target = 30
print(f"\nTarget Element: {target}")
print(element_search(arr,0,target))  # Output: 2
```
