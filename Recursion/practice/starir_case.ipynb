```python
# TC :- O(3^n)
# SC :- O(n)  (recursion stack space)

def count_ways(n:int) -> int:
    # Base cases
    if n == 0:
        return 1
    if n < 0:
        return 0
    
    # Recursive case
    return count_ways(n - 1) + count_ways(n - 2) + count_ways(n - 3)

print(count_ways(4))
```

### `Optimization` : As we can observe same thing is getting calculated multiple times so we will optimize it.

```python
# Time: O(n)
# Space: O(n) (memo + recursion stack)

def count_ways_memo(n:int, memo={}) -> int:
    # Base Cases
    if n == 0:
        return 1
    if n < 0:
        return 0
    
    # Check if result is already computed
    if n in memo:   
        return memo[n]
    print(memo)
    
    # Recursive case with memoization
    memo[n] = (
        count_ways_memo(n - 1, memo) +
        count_ways_memo(n - 2, memo) +
        count_ways_memo(n - 3, memo)
    )
    return memo[n]
```

```python
n = int(input("Enter number of stairs: "))
print("Number of ways:", count_ways(n))
```
