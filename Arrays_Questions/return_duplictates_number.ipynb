```python
nums = [3,1,3,4,2]
nums.sort()
result = -1

for i in range(1, len(nums)):
    if nums[i] == nums[i-1]:
        result = nums[i]
        break
print(result)
```

```python
# TODO: Slow And Fast Pointer Approach
slow, fast = 0, 0
while True:
    slow = nums[slow]
    fast = nums[nums[fast]]
    if slow == fast:
        break
slow = 0
while slow != fast:
    slow = nums[slow]
    fast = nums[fast]
print(slow)
```
