```python
def fun(n):
    total = 0
    while n > 0:
        digit = n % 10
        total += digit ** 2
        n //= 10
    return total


n = 2
slow = n
fast = fun(n)

while fast != 1 and slow != fast:
    slow = fun(slow)
    fast = fun(fun(fast))

if fast == 1:
    print("Happy Number")
else:
    print("Not a Happy Number")
```
