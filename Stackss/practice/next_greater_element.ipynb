```python
arr = [4, 8, 5, 2, 25]
res = [-1] * len(arr)
stack = []
for i in range(len(arr) - 1, -1, -1):
    while stack and stack[-1] < arr[i]:
        stack.pop()
    if stack:
        res[i] = stack[-1]
    stack.append(arr[i])
print(res)
```

```python
arr = [4, 8, 5, 2, 25]
res = [0] * len(arr)
stack = []
counter = 0
for i in range(len(arr) - 1, -1, -1):
    while stack and stack[-1] < arr[i]:
        stack.pop()
    if stack:
        counter += 1
        res[i] = counter
    stack.append(arr[i])
print(res)
```
