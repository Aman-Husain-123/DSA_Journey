```python
s = "()[]{}"
st = []
for ch in s:
    if ch == '(' or ch == '[' or ch == '{':
        st.append(ch)
    else:
        if not st:
            print("False")
            break
        top = st.pop()
        if (ch == ')' and top != '(') or (ch == ']' and top != '[') or (ch == '}' and top != '{'):
            print("False")
            break
else:    
    print("True" if not st else "False")
```
