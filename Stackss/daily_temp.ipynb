```python
temperatures = [30,40,50,60]
n = len(temperatures)
res = [0] * n
stack = []
stack.append(n - 1)

for i in range(n - 1, -1, -1):
    while stack and temperatures[stack[-1]] <= temperatures[i]:
        # print(temperatures[stack[-1]])
        stack.pop()
    
    if stack:
        res[i] = stack[-1] - i
    else:
        res[i] = 0
    
    stack.append(i)
print(res)
```
