```python
arr = [4,1,2,5,3]
lenn = len(arr)
res = [-1]*lenn

for i in range(lenn):
    for j in range(i,-1,-1):
        if arr[j]>arr[i]:
            res[i]=arr[j]
            break
print(res)
```

```python
# Optimal Approach using stack
arr = [10,1,2,5,3]
lenn = len(arr)
res = [-1]*lenn
stack = []

for i in range(lenn-1,-1,-1):
    while stack and arr[stack[-1]]<arr[i]:
        index = stack.pop()
        res[index]=arr[i]
    stack.append(i)
print(res)
```

```python
arr = [10, 1, 2, 5, 3]
n = len(arr)
res = [-1] * n
stack = []

for i in range(n):
    while stack and stack[-1] <= arr[i]:
        stack.pop()

    if stack:
        res[i] = stack[-1]

    stack.append(arr[i])

print(res)
```
