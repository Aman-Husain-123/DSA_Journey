```python
temperatures = [73,74,75,71,69,72,76,73]
n = len(temperatures)
res = [-1] * n
stack = []
stack.append(n - 1)

for i in range(n-1,-1,-1):
    while stack and temperatures[stack[-1]] <= temperatures[i]:
        stack.pop()

        if stack:
            res[i] = stack[-1] - i
        else:
            res[i] = 0

        stack.append(i)
print(res)
```
