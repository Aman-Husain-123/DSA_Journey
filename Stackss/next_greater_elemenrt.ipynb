```python
### `Naive Approach`
```

```python
nums = [800, -2, 16, 3, 100]
n = len(nums)
result = [-1] * n

for i in range(n):
    # search right side
    for j in range(i + 1, n):
        if nums[j] > nums[i]:
            result[i] = nums[j]
            break

    # if not found, search left side
    if result[i] == -1:
        for j in range(0, i):
            if nums[j] > nums[i]:
                result[i] = nums[j]
                break

print(result)
```

### Better Approach

```python
nums = [1,2,1]
n = len(nums)
result = [-1] * n
stack = []

for i in range(n-1,-1,-1):
    stack.append(nums[i])

for j in range(n-1,-1,-1):
    while stack and stack[-1] <= nums[j]:
        stack.pop()

    if stack:
        result[j] = stack[-1]
        
    stack.append(nums[j])
print(result)
```
