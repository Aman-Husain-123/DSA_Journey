```python
arr = [10, 5, 2, 7, 1, -10]
k = 15
presummap = {}
sum = 0
maxLen = 0

for i in range(len(arr)):
    sum += arr[i]
    if sum == k:
        maxLen = max(maxLen,i+1)
    
    rem = sum - k
    if rem in presummap:
        length = i - presummap[rem]
        max_len = max(max_len, length)
    

    # Store prefix sum only if it is not already present
        # (to ensure longest subarray)
    if sum not in presummap:
        presummap[sum] = i

```
