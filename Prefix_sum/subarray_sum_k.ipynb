```python
nums = [1,1,1]
k = 2
arr = []
count = 0
for i in range((len(nums))):
    curr_sum = 0
    for j in range(1,(len(nums))):
        curr_sum += nums[j]
        print(f"{curr_sum} + {nums[j]} = ",curr_sum)

        if curr_sum == k:
            arr.append(nums[j])
            count += 1

print(count)
print(arr)
```
