```python
nums = [1,7,3,6,5,6]
pref = [0] * len(nums)
suff = [0] * len(nums)

for i in range(len(nums)):
    pref[i] = nums[i - 1] + pref[i-1]

for i in range(len(nums)-2, -1, -1):
    suff[i] = nums[i + 1] + suff[i+1]

for i in range(len(nums)):
    if pref[i] == suff[i]:
        print(i)
```

```python
nums = [1,7,3,6,5,6]
summ = sum(nums)
left = 0
ans = -1

for i in range(len(nums)):
    right = summ - left - nums[i]
    if left == right:
        ans = i
        break
    left += nums[i]
print(ans)
```
