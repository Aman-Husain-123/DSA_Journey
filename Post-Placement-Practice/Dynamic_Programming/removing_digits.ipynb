```python
def get_digits(n):
    res = []
    while n > 0:
        if n % 10 != 0:
            res.append(n % 10)
        n //= 10
    
    return res

def f(n,dp):
    if n == 0:
        return 0
    
    if n <= 9:
        return 1
    
    if dp[n] != -1:
        return dp[n]
    
    d = get_digits(n)

    res = float('inf')
    
    for i in range(len(d)):
        res = min(res,f(n-d[i],dp))
    
    dp[n] = 1 + res
    return dp[n]
```

```python
n = 27
dp = [-1] * (n + 1)
f(n,dp)
```

### Bottom Up Approach

```python
def fbu(num):
    dp = [-1] * max(10, num + 1)
    dp[0] = 0
    for i in range(1, 10):
        dp[i] = 1
    for n in range(10, num + 1):
        # n -> state
        # calc dp[n]
        d = get_digits(n)
        result = float('inf')
        for i in range(len(d)):
            result = min(result, dp[n - d[i]])
        dp[n] = 1 + result
    return dp[num]

n = 10
fbu(n)
```
