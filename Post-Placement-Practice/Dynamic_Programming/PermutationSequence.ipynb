```python
class Solution:
    def Permute(self,v,ans,original):
        if len(original) == 0:
            v.append(ans)
            return
        for i in range(len(original)):
            ch = original[i]
            left = original[0:i]
            right = original[i+1:]
            self.Permute(v,ans+ch,left+right)
        
    def getPermutation(self, n: int, k: int) -> str:
        strr = ""
        for i in range(1, n + 1):
            strr += str(i)
        
        v = []
        self.Permute(v,"",strr)
        return v[k-1]

n = 3
k = 3
obj = Solution()
obj.getPermutation(n, k)
```
