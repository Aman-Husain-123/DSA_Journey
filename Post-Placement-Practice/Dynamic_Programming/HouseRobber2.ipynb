```python
# Bottom-Up DP for House Robber I
def rob(nums):
    n = len(nums)

    if n == 0:
        return 0

    if n == 1:
        return nums[0]

    dp = [-1] * n
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])

    for i in range(2, n):
        dp[i] = max(nums[i] + dp[i - 2], dp[i - 1])

    return dp[n - 1]


# House Robber II (Circular Houses)
def robCircular(nums):
    n = len(nums)

    if n == 0:
        return 0

    if n == 1:
        return nums[0]

    # Case 1: Exclude the last house
    money1 = rob(nums[:-1])

    # Case 2: Exclude the first house
    money2 = rob(nums[1:])

    return max(money1, money2)


# Driver Code
nums = [2, 3, 2]
print(robCircular(nums))  # Output: 3

nums = [1, 2, 3, 1]
print(robCircular(nums))  # Output: 4

nums = [1, 2, 3]
print(robCircular(nums))  # Output: 3
```
