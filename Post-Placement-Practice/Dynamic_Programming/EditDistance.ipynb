```python
# Recursion
class Solution:
    def minSteps(self,i,j,a,b):
        if i == -1:
            return j + 1
        if j == -1:
            return i + 1

        if a[i] == b[j]:
            return self.minSteps(i-1,j-1,a,b)
        else:
            dell = self.minSteps(i-1,j,a,b)
            inst = self.minSteps(i,j-1,a,b)
            rep = self.minSteps(i-1,j-1,a,b)
            return 1 + min(dell,min(inst,rep))

    def minDistance(self, word1: str, word2: str) -> int:
        a = word1
        b = word2

        m = len(a)
        n = len(b)

        return self.minSteps(m-1,n-1,a,b)


word1 = "horse"
word2 = "ros"
obj = Solution()
obj.minDistance(word1,word2)



# Insert not required
# if a < b then insert is required
```

```python
# Recursion
class Solution:
    def minSteps(self,i,j,a,b,dp):
        if i == -1:
            return j + 1
        if j == -1:
            return i + 1

        if dp[i][j] != -1:
            return dp[i][j]

        if a[i] == b[j]:
            dp[i][j] = self.minSteps(i-1,j-1,a,b,dp)
            return self.minSteps(i-1,j-1,a,b,dp)
        else:
            dell = self.minSteps(i-1,j,a,b,dp)
            inst = self.minSteps(i,j-1,a,b,dp)
            rep = self.minSteps(i-1,j-1,a,b,dp)
            
            dp[i][j] = 1 + min(dell,min(inst,rep))

            return dp[i][j]

    def minDistance(self, word1: str, word2: str) -> int:
        a = word1
        b = word2

        m = len(a)
        n = len(b)
        dp = [[-1] * n for _ in range(m)]
        return self.minSteps(m-1,n-1,a,b,dp)


word1 = "horse"
word2 = "ros"
obj = Solution()
obj.minDistance(word1,word2)



# Insert not required
# if a < b then insert is required
```
