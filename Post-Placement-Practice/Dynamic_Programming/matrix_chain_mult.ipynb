# `Recursion`

```python
# Minimum Costing
def mcm(i,j,arr):
    if i == j:
        return 0
        
    minCost = float('inf')
    
    for k in range(i,j):
        x = arr[i][0] * arr[j][1] * arr[k][1]
        totalcost = mcm(i,k,arr) + mcm(k + 1,j,arr) + x
        minCost = min(minCost,totalcost)
    return minCost

arr = [[1,2],[2,3],[3,4],[4,2]]
n = len(arr)
mcm(0,n-1,arr)
```

```python
# Maximum Costing
def mcm(i,j,arr):
    if i == j:
        return 0
        
    minCost = float('-inf')
    
    for k in range(i,j):
        x = arr[i][0] * arr[j][1] * arr[k][1]
        totalcost = mcm(i,k,arr) + mcm(k + 1,j,arr) + x
        minCost = max(minCost,totalcost)
    return minCost

arr = [[1,2],[2,3],[3,4]]
n = len(arr)
mcm(0,n-1,arr)
```

# `1-D Array`

```python
# Minimum Costing
def mcm(i,j,arr):
    if i == j:
        return 0
        
    minCost = float('inf')
    
    for k in range(i,j):
        x = arr[i] * arr[j+1] * arr[k+1]
        totalcost = mcm(i,k,arr) + mcm(k + 1,j,arr) + x
        minCost = min(minCost,totalcost)
    return minCost

arr = [1,2,3,4]
n = len(arr)
mcm(0,n-2,arr)
```

# `Memoization`

```python
# Minimum Costing
def mcm(i,j,arr,dp):
    if i == j:
        return 0

    if dp[i][j] != -1:
        return dp[i][j]

    minCost = float('inf')
    
    for k in range(i,j):
        x = arr[i] * arr[j+1] * arr[k+1]
        totalcost = mcm(i,k,arr,dp) + mcm(k + 1,j,arr,dp) + x
        minCost = min(minCost,totalcost)
    
    dp[i][j] = minCost
    return dp[i][j]

arr = [1,2,3,4]
n = len(arr)
dp = [[-1] * (n-1) for _ in range(n-1)]
mcm(0,n-2,arr,dp)
```

# `Tabulation`

```python
# Minimum Costing
def mcm(i,j,arr,dp):
    if i == j:
        return 0

    if dp[i][j] != -1:
        return dp[i][j]

    minCost = float('inf')

    for k in range(i,j):
        x = arr[i] * arr[j+1] * arr[k+1]
        totalcost = mcm(i,k,arr,dp) + mcm(k + 1,j,arr,dp) + x
        minCost = min(minCost,totalcost)
    
    dp[i][j] = minCost
    return dp[i][j]

arr = [1,2,3,4]
n = len(arr)
dp = [[-1] * (n-1) for _ in range(n-1)]
mcm(0,n-2,arr,dp)
```

# `Another Solution`

```python
# Memoization

from typing import List
def f(i:int, j:int, arr:List[int]) -> int:
    if i == j or i + 1 == j:
        return 0
    
    if dp[i][j] != -1:
        return dp[i][j]
    
    ans = float("inf")
    for k in range(i + 1,j):
        ans = min(ans,f(i,k,arr) + f(k,j,arr) + (arr[i] * arr[k] * arr[j]))
    
    dp[i][j] = ans
    return dp[i][j]

n =  [1, 2, 3, 4, 3]

dp = [[-1] * len(n) for _ in range(len(n))]
f(0,len(n)-1,n)
```

```python
# Tabulation
def matrix_chain(arr):
    n = len(arr)

    # Create DP table
    dp = [[0] * n for _ in range(n)]

    # Length of current range
    for length in range(3, n + 1):

        for i in range(n - length + 1):

            j = i + length - 1

            dp[i][j] = float('inf')

            # Try every possible partition
            for k in range(i + 1, j):

                cost = (
                    dp[i][k]
                    + dp[k][j]
                    + arr[i] * arr[k] * arr[j]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[0][n - 1]


arr = [10, 20, 30, 40, 30]

print(matrix_chain(arr))
```
