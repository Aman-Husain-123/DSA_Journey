```python
# Recursion
class Solution:
    def f(self,n):
        if n == 0 or n == 1:
            return 1
        
        if n == 2:
            return 2
        
        summ = 0
        for k in range(1,n+1):
            # k -> Potential root
            summ += self.f(k-1) * self.f(n-k)
        
        return summ
        
    def numTrees(self, n: int) -> int:
        return self.f(n)
```

```python
# DP
class Solution:
    def f(self,n,dp):
        if n == 0 or n == 1:
            return 1
        
        if n == 2:
            return 2 
        
        if dp[n] != -1:
            return dp[n]
        
        summ = 0
        for k in range(1,n+1):
            # k -> Potential root
            summ += self.f(k-1,dp) * self.f(n-k,dp)
        
        dp[n] = summ
        return dp[n]

    def numTrees(self, n: int) -> int:
        dp = [-1] * 30 
        return self.f(n,dp)
```
