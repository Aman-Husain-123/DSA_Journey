```python
class Solution:

    def minCost(self, n, cap, ass):

        dp = [[-1] * (n + 1) for _ in range(n)]

        def f(i, x):

            # All pilots assigned
            if i == n:
                return 0

            # Already calculated
            if dp[i][x] != -1:
                return dp[i][x]

            # If x == 0, must choose assistant
            if x == 0:
                dp[i][x] = ass[i] + f(i + 1, x + 1)

            # If x == n-i, must choose captain
            elif x == n - i:
                dp[i][x] = cap[i] + f(i + 1, x - 1)

            # Both choices are possible
            else:

                assistant = ass[i] + f(i + 1, x + 1)

                captain = cap[i] + f(i + 1, x - 1)

                dp[i][x] = min(assistant, captain)

            return dp[i][x]

        return f(0, 0)


# Driver Code
if __name__ == "__main__":

    n = int(input())

    cap = []
    ass = []

    for i in range(n):
        c, a = map(int, input().split())
        cap.append(c)
        ass.append(a)

    sol = Solution()

    print(sol.minCost(n, cap, ass))
```
