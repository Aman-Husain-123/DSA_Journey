```python
# TC :- 2^N
# SC :- O(n*t)
def subset(i, summ, arr, target):
    if i == len(arr):
        if summ == target:
            return True
        else:
            return False
    skip = subset(i + 1, summ, arr, target)

    # When elements are neg then below if condition shouldnt be included
    if summ + arr[i] > target:
        return skip
    pick = subset(i + 1, summ , arr, target - arr[i])

    return pick or skip
```

```python
arr = [8,5,2,4]
target = 9

subset(0,0,arr,target)
```

# `DP` 

```python
class Solution:

    def subset(self, i, arr, target, dp):

        if i == len(arr):
            return target == 0

        if dp[i][target] != -1:
            return dp[i][target] == 1

        ans = False

        # Skip the current element
        skip = self.subset(i + 1, arr, target, dp)

        if target - arr[i] < 0:
            ans = skip
        else:
            # Pick the current element
            pick = self.subset(i + 1, arr, target - arr[i], dp)
            ans = pick or skip

        if ans:
            dp[i][target] = 1
        else:
            dp[i][target] = 0

        return ans

    def isSubsetSum(self, arr, target):

        dp = [[-1] * (target + 1) for _ in range(len(arr) + 1)]

        return self.subset(0, arr, target, dp)


# Example
arr = [3, 34, 4, 12, 5, 2]
target = 9

obj = Solution()
print(obj.isSubsetSum(arr, target))
```
