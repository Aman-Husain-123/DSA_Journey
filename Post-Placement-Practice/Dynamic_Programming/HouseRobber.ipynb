# `Recursion`

```python
def amount(nums,i):
    if i >= len(nums):
        return 0
    
    take = nums[i] + amount(nums,i + 2)
    skipp = amount(nums,i+1)

    return max(take,skipp)

def rob(nums):
    return amount(nums,0)

nums = [1,2,3,1]
rob(nums)
```

# `Bottom-Up Approach`

```python
def fbu(arr, dp):
    n = len(arr)

    if n == 1:
        return arr[0]

    dp[n - 1] = arr[n - 1]
    dp[n - 2] = max(arr[n - 2], arr[n - 1])

    for i in range(n - 3, -1, -1):
        dp[i] = max(arr[i] + dp[i + 2], dp[i + 1])

    return dp[0]


def rob(nums):
    n = len(nums)
    dp = [-1] * n
    return fbu(nums, dp)


nums = [1, 2, 3, 1]
print(rob(nums))
```
