```python
from typing import List
class Solution:

    def f(self, n, i, j, k, dp):

        # Knight went outside the board
        if i < 0 or j < 0 or i >= n or j >= n:
            return 0.0

        # No more moves
        if k == 0:
            return 1.0

        # Already calculated
        if dp[i][j][k] != -1.0:
            return dp[i][j][k]

        ans = 0.0

        # 8 possible knight moves
        ans += self.f(n, i + 1, j + 2, k - 1, dp) * 0.125
        ans += self.f(n, i + 2, j + 1, k - 1, dp) * 0.125
        ans += self.f(n, i + 1, j - 2, k - 1, dp) * 0.125
        ans += self.f(n, i + 2, j - 1, k - 1, dp) * 0.125

        ans += self.f(n, i - 1, j + 2, k - 1, dp) * 0.125
        ans += self.f(n, i - 2, j + 1, k - 1, dp) * 0.125
        ans += self.f(n, i - 1, j - 2, k - 1, dp) * 0.125
        ans += self.f(n, i - 2, j - 1, k - 1, dp) * 0.125

        dp[i][j][k] = ans

        return dp[i][j][k]

    def knightProbability(self, n: int, k: int, row: int, column: int) -> float:

        # 3D DP
        dp = [[[-1.0 for _ in range(k + 1)]for _ in range(n)]for _ in range(n)]
        return self.f(n, row, column, k, dp)

n = 3
k = 2
row = 0 
column = 0
obj = Solution()
obj.knightProbability(n,k,row,column)
```

```python
class Solution:

    def knightProbability(self, n: int, k: int, row: int, column: int) -> float:

        # dp[row][column][moves]
        dp = [
            [
                [-1.0] * (k + 1)
                for _ in range(n)
            ]
            for _ in range(n)
        ]

        dx = [1, 2, 1, 2, -1, -2, -1, -2]
        dy = [2, 1, -2, -1, 2, 1, -2, -1]

        def f(i, j, k):

            # Outside the chessboard
            if i < 0 or j < 0 or i >= n or j >= n:
                return 0.0

            # No moves remaining
            if k == 0:
                return 1.0

            # Already calculated
            if dp[i][j][k] != -1.0:
                return dp[i][j][k]

            ans = 0.0

            # Try all 8 knight moves
            for m in range(8):
                ans += f(i + dx[m],j + dy[m],k - 1) * 0.125

            dp[i][j][k] = ans

            return ans

        return f(row, column, k)


# Test case
n = 3
k = 2
row = 0
column = 0

obj = Solution()

print(obj.knightProbability(n, k, row, column))
```
