```python
# TC :- O(N Log N)

def fractionalKnapsack(profit,weights,W):
    arr = []
    n = len(profit)

    # Store (profit, weight)
    for i in range(n):
        arr.append((profit[i] , weights[i]))

    # Sort the profit/weight ratio in desc order
    arr.sort(key=lambda x:x[0] // x[1],reverse = True)

    result = 0.0

    for p,wt in arr:
        if wt <= W:
            result += p
            W -= wt
        else:
            result += (p/wt) * W
            W = 0
            break
    return result

val = [60, 100, 120]
wt = [10, 20, 30]
capacity = 50    
fractionalKnapsack(val,wt,capacity)
```
