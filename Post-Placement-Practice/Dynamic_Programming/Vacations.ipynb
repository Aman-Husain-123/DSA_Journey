```python
n = int(input())

# dp[i][0] -> Maximum happiness if activity A is chosen on day i
# dp[i][1] -> Maximum happiness if activity B is chosen on day i
# dp[i][2] -> Maximum happiness if activity C is chosen on day i
dp = [[0] * 3 for _ in range(n)]

# Base case (Day 0)
a, b, c = map(int, input().split())
dp[0][0] = a
dp[0][1] = b
dp[0][2] = c

# Fill DP table
for i in range(1, n):
    # Happiness values for the ith day
    a, b, c = map(int, input().split())

    # Choose activity A today
    dp[i][0] = a + max(dp[i - 1][1], dp[i - 1][2])

    # Choose activity B today
    dp[i][1] = b + max(dp[i - 1][0], dp[i - 1][2])

    # Choose activity C today
    dp[i][2] = c + max(dp[i - 1][0], dp[i - 1][1])

# Answer
print(max(dp[n - 1][0], dp[n - 1][1], dp[n - 1][2]))
```
