```python
def f(n):
    if n == 1:
        return 0

    if n == 2 or n == 3:
        return 1

    ans = f(n - 1)

    if n % 2 == 0:
        ans = min(ans, f(n // 2))

    if n % 3 == 0:
        ans = min(ans, f(n // 3))

    return 1 + ans


# Driver Code
n = int(input("Enter n: "))
print(f(n))
```

# `Top Down Approach`

```python
def ftd(n, dp):
    if n == 1:
        return 0

    if n == 2 or n == 3:
        return 1

    # Already computed
    if dp[n] != -1:
        return dp[n]

    dp[n] = 1 + min(
        ftd(n - 1, dp),
        ftd(n // 2, dp) if n % 2 == 0 else float('inf'),
        ftd(n // 3, dp) if n % 3 == 0 else float('inf')
    )

    return dp[n]


# Driver Code
n = int(input("Enter n: "))
dp = [-1] * (n + 1)
print(ftd(n, dp))
```

# `Bottom UP`

```python
def fbu(n):
    if n == 1:
        return 0
    if n == 2 or n == 3:
        return 1

    dp = [-1] * (n + 1)

    dp[1] = 0
    dp[2] = dp[3] = 1

    for i in range(4, n + 1):
        dp[i] = 1 + min(
            dp[i - 1],
            dp[i // 2] if i % 2 == 0 else float('inf'),
            dp[i // 3] if i % 3 == 0 else float('inf')
        )

    return dp[n]


# Driver Code
n = int(input("Enter n: "))
print(fbu(n))
```
