```python
k, l, m = map(int, input().split())

# dp[i] = True  -> Winning state
# dp[i] = False -> Losing state
dp = [False] * 1000005

dp[1] = True
dp[k] = True
dp[l] = True

for i in range(2, 1000001):
    if i == k or i == l:
        continue

    dp[i] = not (
        dp[i - 1]
        and (dp[i - k] if i - k >= 1 else True)
        and (dp[i - l] if i - l >= 1 else True)
    )

for _ in range(m):
    n = int(input())

    if dp[n]:
        # Winning state
        print("A", end="")
    else:
        # Losing state
        print("B", end="")

print()
```
