# `Recursion`

```python
class Solution:
    def f(self, i, j, mat):
        if i < 0 or j < 0 or j >= self.m:
            return float('-inf')

        if i == 0:
            return mat[0][j]

        s = mat[i][j] + self.f(i - 1, j, mat)
        ld = mat[i][j] + self.f(i - 1, j - 1, mat)
        rd = mat[i][j] + self.f(i - 1, j + 1, mat)

        return max(s, ld, rd)

    def maximumPath(self, mat):
        self.n = len(mat)
        self.m = len(mat[0])

        ans = float('-inf')

        # Try every column in the last row
        for j in range(self.m):
            ans = max(ans, self.f(self.n - 1, j, mat))

        return ans
```

```python
mat = [[3, 6, 1], [2, 3, 4], [5, 5, 1]]
obj = Solution()
obj.maximumPath(mat)
```

# `Memoization`

```python
class Solution:
    def f(self, i, j, mat, dp):
        if i < 0 or j < 0 or j >= self.m:
            return float('-inf')

        if i == 0:
            return mat[0][j]

        if dp[i][j] != -1:
            return dp[i][j]

        s = mat[i][j] + self.f(i - 1, j, mat, dp)
        ld = mat[i][j] + self.f(i - 1, j - 1, mat, dp)
        rd = mat[i][j] + self.f(i - 1, j + 1, mat, dp)

        dp[i][j] = max(s, ld, rd)
        return dp[i][j]

    def maximumPath(self, mat):
        self.n = len(mat)
        self.m = len(mat[0])

        dp = [[-1] * self.m for _ in range(self.n)]

        ans = float('-inf')

        # Try every column in the last row
        for j in range(self.m):
            ans = max(ans, self.f(self.n - 1, j, mat, dp))

        return ans
```

```python
mat = [[3, 6, 1], [2, 3, 4], [5, 5, 1]]
obj = Solution()
obj.maximumPath(mat)
```

# `Tabulation`

```python
# Tabulation
class Solution:
    def maximumPath(self, mat):
        n = len(mat)
        m = len(mat[0])

        dp = [[0] * m for _ in range(n)]

        # Base case: 1st Row
        for j in range(m):
            dp[0][j] = mat[0][j]
        
        # Fill the dp table
        for i in range(1,n):
            for j in range(m):
                up = mat[i][j] + dp[i-1][j]

                left_diag = mat[i][j]
                if j > 0:
                    left_diag += dp[i-1][j-1]
                else:
                    left_diag += float('-inf')
                right_diag = mat[i][j]

                if j < m - 1:
                    right_diag += dp[i-1][j+1]
                else:
                    right_diag += float('-inf')
                
                dp[i][j] = max(up,left_diag,right_diag)
        
        return max(dp[n-1])
```

```python
mat = [[2,1,3],[6,5,4],[7,8,9]]

obj = Solution()
obj.maximumPath(mat)
```
