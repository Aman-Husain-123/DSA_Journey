```python
def pair(n,dp):
    if n <= 2:
        return n
    
    if dp[n] != -1:
        return dp[n]
    
    dp[n] = pair(n-1,dp) + (n-1) * pair(n-2,dp)

    return dp[n]

n = int(input())
dp = [-1] * (n + 1)
pair(n,dp)
```
