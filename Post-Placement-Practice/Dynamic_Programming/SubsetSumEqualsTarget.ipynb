```python
# Memoization
# TC: O(N * target)
# SC: O(N * target) + O(N) recursion stack



# Memoization TC :- O(N * Target)
# SC :- O(N*Target) + O(N)
def f(ind, target, arr, dp):

    # Base case
    if target == 0:
        return True

    if ind == 0:
        return arr[0] == target

    # Already calculated
    if dp[ind][target] != -1:
        return dp[ind][target]

    # Not take current element
    not_take = f(ind - 1, target, arr, dp)

    # Take current element
    take = False

    if target >= arr[ind]:
        take = f(ind - 1, target - arr[ind], arr, dp)

    # Store result
    dp[ind][target] = take or not_take

    return dp[ind][target]
```

```python
arr = [1, 2, 3, 4]
target = 5

n = len(arr)

dp = [[-1] * (target + 1) for _ in range(n)]

print(f(n - 1, target, arr, dp))
```
