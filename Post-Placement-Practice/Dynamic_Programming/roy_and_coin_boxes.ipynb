```python
n = int(input())

# f[i] = number of coins in box i
f = [0] * (n + 1)

m = int(input())

# l[i] = how many ranges start at i
# r[i] = how many ranges end at i
l = [0] * (n + 1)
r = [0] * (n + 1)

for _ in range(m):
    L, R = map(int, input().split())
    l[L] += 1
    r[R] += 1

# Compute number of coins in each box
f[1] = l[1]

for i in range(2, n + 1):
    f[i] = f[i - 1] + l[i] - r[i - 1]

# c[x] = number of boxes having exactly x coins
c = [0] * 10000005

for i in range(1, n + 1):
    coins = f[i]
    c[coins] += 1

# Convert into suffix sum
# c[x] = number of boxes having at least x coins
for i in range(len(c) - 2, -1, -1):
    c[i] += c[i + 1]

q = int(input())

while q:
    q -= 1
    num = int(input())
    print(c[num])
```
