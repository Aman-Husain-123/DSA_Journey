```python
def f(x, coins, dp):
    if x == 0:
        return 0

    if dp[x] != -1:
        return dp[x]

    result = float('inf')

    for coin in coins:
        if x - coin >= 0:
            result = min(result, f(x - coin, coins, dp))

    if result == float('inf'):
        dp[x] = float('inf')
    else:
        dp[x] = 1 + result

    return dp[x]


# ---------------- Driver Code ----------------
n, x = map(int, input().split())

coins = list(map(int, input().split()))

dp = [-1] * (x + 1)

ans = f(x, coins, dp)

if ans == float('inf'):
    print(-1)
else:
    print(ans)
```
