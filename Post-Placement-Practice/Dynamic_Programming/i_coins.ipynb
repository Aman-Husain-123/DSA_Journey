```python
# All those cases where H is (n + 1) // 2 that would be favourable case
```

```python
class Solution:

    def __init__(self):
        self.dp = [[-1.0] * 3005 for _ in range(3005)]

    def f(self, p, i, x):

        # Base case: we need 0 heads
        if x == 0:
            return 1.0

        # No coins left
        if i == -1:
            return 0.0

        # Already calculated
        if self.dp[i][x] > -0.9:
            return self.dp[i][x]

        # Choose current coin as head
        take = self.f(p, i - 1, x - 1) * p[i]

        # Choose current coin as tail
        skip = self.f(p, i - 1, x) * (1 - p[i])

        self.dp[i][x] = take + skip

        return self.dp[i][x]


# Input
n = int(input())

p = list(map(float, input().split()))

obj = Solution()

answer = obj.f(p, n - 1, (n + 1) // 2)

print(answer)
```
