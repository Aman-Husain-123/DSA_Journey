```python
# TODO: Recursion
from typing import List
class Solution:
    def solve(self, i, j, triangle, n):
        # If at bottom row
        if i == n - 1:
            return triangle[i][j]

        # Compute both downward and diagonal moves
        down = triangle[i][j] + self.solve(i + 1, j, triangle, n)
        diag = triangle[i][j] + self.solve(i + 1, j + 1, triangle, n)

        # Store and return min path sum
        return min(down, diag)

    def minimumTotal(self, triangle: List[List[int]]) -> int:
        n = len(triangle)
        dp = [[-1 for _ in range(n)] for _ in range(n)]
        return self.solve(0, 0, triangle, n)

triangle = [[2],[3,4],[6,5,7],[4,1,8,3]]
obj = Solution()
obj.minimumTotal(triangle)
```
