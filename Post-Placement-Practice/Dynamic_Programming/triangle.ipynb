```python
class Solution:
    def solve(self, i, j, triangle, n, dp):
        # If value already computed
        if dp[i][j] != -1:
            return dp[i][j]

        # If at bottom row
        if i == n - 1:
            return triangle[i][j]

        # Compute both downward and diagonal moves
        down = triangle[i][j] + self.solve(i + 1, j, triangle, n, dp)
        diag = triangle[i][j] + self.solve(i + 1, j + 1, triangle, n, dp)

        # Store and return min path sum
        dp[i][j] = min(down, diag)
        return dp[i][j]

    def minimumTotal(self, triangle: List[List[int]]) -> int:
        n = len(triangle)
        dp = [[-1 for _ in range(n)] for _ in range(n)]
        return self.solve(0, 0, triangle, n, dp)

triangle = [[2],[3,4],[6,5,7],[4,1,8,3]]
obj = Solution()
obj.minimumTotal(triangle)
```

```python
from typing import List
def minimumTotal(triangle: List[List[int]]) -> int:
        n = len(triangle)
        dp = [[-1 for _ in range(n)] for _ in range(n)]
        
        for j in range(0,n):
            dp[n-1][j] = triangle[n-1][j]
        
        for i in range(n-2,-1,-1):
            for j in range(i + 1):
                down = triangle[i][j] + dp[i+1][j]
                diag = triangle[i][j] + dp[i+1][j+1]

                dp[i][j] = min(down,diag)
        return dp[0][0]

triangle = [[2],[3,4],[6,5,7],[4,1,8,3]]
minimumTotal(triangle)
```

# `Space Optimization`

```python
from typing import List
class Solution:
    def minimumTotal(self, triangle: List[List[int]]) -> int:
        n = len(triangle)

        # # Last row forms the initial DP array
        prev = [-1] * n
        for j in range(n):
            prev[j] = triangle[n-1][j]
        
        # Move upwards from second-last row
        for i in range(n-2,-1,-1):
            curr = [-1] * (i + 1)
            for j in range(i+1):
                # Minimum path sum for current cell
                down = triangle[i][j] + prev[j]
                diag = triangle[i][j] + prev[j+1]

                curr[j] = min(down,diag)
            prev = curr
    
        return prev[0]

triangle = [[2],[3,4],[6,5,7],[4,1,8,3]]
obj = Solution()
obj.minimumTotal(triangle)
```
