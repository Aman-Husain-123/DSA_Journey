# `Recursive Approach`

```python
# n = Number of dice
# k = Number of faces of dice(1 --> k)   e.g :- k = 6 (faces = 1 --> 6)
# t = target sum
def f(n,k,t):
    if n == 0 and t == 0:
        return 1
    if n == 0:
        return 0
    
    summ = 0
    for v in range(1,k+1):
        summ += f(n - 1,k,t-v)
    
    return summ
```

```python
def numRollsTarget(n,k,target):
    return f(n,k,target)
```

```python
n = 1
k = 6
target = 3

numRollsTarget(n,k,target)
```

# `Memoization`

```python
# There are 2 variable "n" and "t" by using it we identify the subproblem and they are changing
MOD = 1000000007

def f(n, k, t, dp):
    if n == 0 and t == 0:
        return 1
    if n == 0:
        return 0

    if dp[n][t] != -1:
        return dp[n][t]

    summ = 0
    for v in range(1, k + 1):
        if (t - v) < 0:
            continue
        summ = (summ + f(n - 1, k, t - v, dp)) % MOD

    dp[n][t] = summ
    return summ


def numRollsTarget(n, k, target):
    dp = [[-1] * (target + 1) for _ in range(n + 1)]
    return f(n, k, target, dp)


n = 1
k = 6
target = 3

print(numRollsTarget(n, k, target))
```
