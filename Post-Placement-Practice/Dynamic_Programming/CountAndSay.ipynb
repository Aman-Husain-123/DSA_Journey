```python
class Solution:
    def countAndSay(self, n: int) -> str:
        if n == 1:
            return "1"
        
        s = self.countAndSay(n-1)

        result = ""

        freq = 1
        ch = s[0]

        for i in range(1,len(s)):
            dh = s[i]

            if ch == dh:
                freq += 1
            
            else:
                result += str(freq) + ch

                freq = 1
                ch = dh
        
        # Add the last group
        result += str(freq) + ch
        return result
n = 4
solution = Solution()
print(solution.countAndSay(n))  # Output: "1211"
```
