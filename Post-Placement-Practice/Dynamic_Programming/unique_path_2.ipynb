```python
class Solution:
    def uniquePathsWithObstacles(self, obstacleGrid):
        m = len(obstacleGrid)
        n = len(obstacleGrid[0])

        if obstacleGrid[0][0] == 1:
            return 0

        dp = [[0] * n for _ in range(m)]

        for i in range(m):
            for j in range(n):
                
                # Obstacle find
                if obstacleGrid[i][j] == 1:
                    dp[i][j] = 0

                # Starting cell
                elif i == 0 and j == 0:
                    dp[i][j] = 1

                # Row
                elif i == 0:
                    dp[i][j] = dp[i][j - 1]

                # Col
                elif j == 0:
                    dp[i][j] = dp[i - 1][j]

                # top + left
                else:
                    dp[i][j] = dp[i - 1][j] + dp[i][j - 1]

        return dp[m - 1][n - 1]
```

```python
obstacleGrid = [[0,0,0],[0,1,0],[0,0,0]]
obj = Solution()
obj.uniquePathsWithObstacles(obstacleGrid)
```
