```python
class Solution:
    def longCommSubstr(self, s1, s2):
        # code here
        m = len(s1)
        n = len(s2)
        
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        
        maxx = 0
        
        for i in range(1,m+1):
            for j in range(1,n+1):
                if s1[i-1] == s2[j-1]:
                    dp[i][j] = 1 + dp[i-1][j-1]
                else:
                    dp[i][j] = 0
                maxx = max(maxx,dp[i][j])
            
        return maxx
```

```python
s1 = "ABCDGH"
s2 = "ACDGHR"

obj = Solution()
obj.longCommSubstr(s1,s2)
```

# `Space Optimization`

```python
class Solution:
    def longCommSubstr(self, s1, s2):
        m = len(s1)
        n = len(s2)

        prev = [0] * (n + 1)
        curr = [0] * (n + 1)

        ans = 0

        for i in range(1, m + 1):
            for j in range(1, n + 1):

                if s1[i - 1] == s2[j - 1]:
                    curr[j] = 1 + prev[j - 1]
                else:
                    curr[j] = 0

                ans = max(ans, curr[j])

            prev[:] = curr[:]
            curr[:] = [0] * (n + 1)

        return ans
```

```python
s1 = "ABCDGH"
s2 = "ACDGHR"

obj = Solution()
obj.longCommSubstr(s1,s2)
```
