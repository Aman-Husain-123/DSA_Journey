```python
from typing import List

class Solution:
    def solve(self, i, j1, j2, n, m, grid, dp):
        # TC :- O(N*M*M) * 9
        # SC :- O(N*M*M) * O(N)     ---->  Auxilary Stack Space

        # Out of boundary
        if j1 < 0 or j1 >= m or j2 < 0 or j2 >= m:
            return -10**9

        # Base case: last row
        if i == n - 1:
            if j1 == j2:
                return grid[i][j1]
            else:
                return grid[i][j1] + grid[i][j2]

        # DP check
        if dp[i][j1][j2] != -1:
            return dp[i][j1][j2]

        maxi = -10**9

        # Try all 9 combinations
        for dj1 in [-1, 0, 1]:
            for dj2 in [-1, 0, 1]:

                if j1 == j2:
                    maxi = max(
                        maxi,
                        grid[i][j1] +
                        self.solve(
                            i + 1,
                            j1 + dj1,
                            j2 + dj2,
                            n, m, grid, dp
                        )
                    )

                else:
                    maxi = max(
                        maxi,
                        grid[i][j1] + grid[i][j2] +
                        self.solve(
                            i + 1,
                            j1 + dj1,
                            j2 + dj2,
                            n, m, grid, dp
                        )
                    )

        # Store answer
        dp[i][j1][j2] = maxi

        return dp[i][j1][j2]

    def cherryPickup(self, grid: List[List[int]]) -> int:

        n = len(grid)
        m = len(grid[0])

        # dp[row][robot1_col][robot2_col]
        dp = [[[-1] * m for _ in range(m)] for _ in range(n)]

        return self.solve(0, 0, m - 1, n, m, grid, dp)
```

```python
grid = [[3,1,1],[2,5,1],[1,5,5],[2,1,1]]
obj = Solution()
obj.cherryPickup(grid)
```
