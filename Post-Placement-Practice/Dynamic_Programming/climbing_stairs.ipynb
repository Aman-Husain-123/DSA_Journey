```python
# TODO: Recursive Approach
def f(n,dp):
    if n == 0 or n == 1:
        return 1
    
    if dp[n] != -1:
        return dp[n]
    
    dp[n] = f(n-1,dp) + f(n-2,dp)
    
    return dp[n]

n = 5
dp = [-1] * (n+1)
f(n,dp)
```

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        dp = {}

        def fun(i):
            # Base case: reached exactly the nth stair
            if i == n:
                return 1

            # Base case: crossed the nth stair
            if i > n:
                return 0

            # If already computed, return from dp
            if i in dp:
                return dp[i]

            # Recursive calls
            a1 = fun(i + 1)
            a2 = fun(i + 2)

            # Store the result
            dp[i] = a1 + a2

            print(dp)

            return dp[i]

        return fun(0)


# Driver Code
if __name__ == "__main__":
    n = int(input("Enter the number of stairs: "))

    sol = Solution()
    ans = sol.climbStairs(n)

    print("Number of ways to climb", n, "stairs:", ans)
```

```python
# Tabulation Approach
class Solution:
    def climbStairs(self, n: int) -> int:
        # TODO: Tabulation Code
        if  n == 1 or n == 2:
            return n
        
        dp = [-1] * (n + 1)

        dp[1] = 1
        dp[2] = 2

        for i in range(3,n+1):
            dp[i] = dp[i - 1] + dp[i-2]
        
        return dp[n]
n = 3
obj = Solution()
obj.climbStairs(n)
```

```python
# TODO: Space Optimization
class Solution:
    def climbStairs(self, n: int) -> int:
        # TODO: Space optimized code
        if n == 1 or n == 2:
            return n
        
        prev2 = 1
        prev1 = 2

        res = prev1

        for i in range(3,n+1):
            res = prev1 + prev2
            prev2 = prev1
            prev1 = res
        
        return res

# Driver Code
if __name__ == "__main__":
    n = int(input("Enter the number of stairs: "))

    sol = Solution()
    ans = sol.climbStairs(n)

    print("Number of ways to climb", n, "stairs:", ans)
```
