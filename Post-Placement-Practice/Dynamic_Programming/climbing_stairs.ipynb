```python
class Solution:
    def climbStairs(self, n: int) -> int:
        dp = {}

        def fun(i):
            # Base case: reached exactly the nth stair
            if i == n:
                return 1

            # Base case: crossed the nth stair
            if i > n:
                return 0

            # If already computed, return from dp
            if i in dp:
                return dp[i]

            # Recursive calls
            a1 = fun(i + 1)
            a2 = fun(i + 2)

            # Store the result
            dp[i] = a1 + a2

            print(dp)

            return dp[i]

        return fun(0)


# Driver Code
if __name__ == "__main__":
    n = int(input("Enter the number of stairs: "))

    sol = Solution()
    ans = sol.climbStairs(n)

    print("Number of ways to climb", n, "stairs:", ans)
```
