```python
def derangeCount(n):
    if n == 1:
        return 0
    if n == 2:
        return 1
    
    return (n - 1) * (derangeCount(n - 1) + derangeCount(n - 2))
    
if __name__ == "__main__":
    n = 5
    print(derangeCount(n))
```

# `Memoization`

```python
def derangeCount(n,dp):
    if n == 1:
        return 0
    if n == 2:
        return 1
    
    if dp[n] != -1:
        return dp[n]
    
    dp[n] = (n - 1) * (derangeCount(n - 1,dp) + derangeCount(n - 2,dp))
    return dp[n]
    
if __name__ == "__main__":
    n = 5
    dp = [-1] * (n + 1)
    print(derangeCount(n,dp))
```

# `Tabulation`

```python
def derangeCount(n):
  
    # Create a DP array to store results
    dp = [0] * (n + 1)

    # Base cases
    dp[1] = 0
    dp[2] = 1

    # Fill the DP array using the 
    # recursive relation
    for i in range(3, n + 1):
        dp[i] = (i - 1) * (dp[i - 1] + dp[i - 2])
    
    return dp[n]

if __name__ == "__main__":
    n = 5
    print(derangeCount(n))
```
