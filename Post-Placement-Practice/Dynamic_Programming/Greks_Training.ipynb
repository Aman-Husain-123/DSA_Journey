```python
# O(3^n)
# O(n)
# TODO: Recursion
class Solution:
    def solve(self,day,last,arr):
        # Base case: Day 0, pick max task != last
        if day == 0:
            maxi = 0
            for i in range(0, 3):
                if last != i:
                    maxi = max(maxi, arr[day][i])
            return maxi
            
        # For current day, try all valid tasks
        maxi = 0
        for i in range(0, 3):
            if last != i:
                maxi = max(maxi, arr[day][i] + self.solve(day - 1, i, arr))
        return maxi
        
    def maximumPoints(self, mat):
        n = len(mat)
        # Start from last day with no previous task (last=3)
        return self.solve(n - 1, 3, mat)

mat = [[1, 2, 5],
      [3, 1, 1], 
      [3, 3, 3]]

obj = Solution()
obj.maximumPoints(mat)
```

```python
# Memoization
# TODO: Recursion
class Solution:
    def solve(self,day,last,arr,dp):

        # Check memo
        if dp[day][last] != -1:
            return dp[day][last]

        # Base case: Day 0, pick max task != last
        if day == 0:
            maxi = 0
            for i in range(0, 3):
                if last != i:
                    maxi = max(maxi, arr[day][i])
            return maxi
            
        # For current day, try all valid tasks
        maxi = 0
        for i in range(0, 3):
            if last != i:
                maxi = max(maxi, arr[day][i] + self.solve(day - 1, i, arr,dp))

        # Store in memo
        dp[day][last] = maxi
        return dp[day][last]
        
    def maximumPoints(self, mat):
        n = len(mat)
        # Start from last day with no previous task (last=3)

        dp = [[-1 for _ in range(4)] for _ in range(n)]
        return self.solve(n - 1, 3, mat,dp)

mat = [[1, 2, 5],
      [3, 1, 1], 
      [3, 3, 3]]

n = len(mat)
obj = Solution()
obj.maximumPoints(mat)
```
