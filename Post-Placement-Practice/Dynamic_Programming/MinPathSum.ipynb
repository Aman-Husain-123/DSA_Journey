```python
# TODO: Tabulation
from typing import List

class Solution:
    def minPathSum(self, arr: List[List[int]]) -> int:
        m = len(arr)
        n = len(arr[0])

        dp = [[0] * n for _ in range(m)]

        for i in range(m):
            for j in range(n):
                if i == 0 and j == 0:
                    dp[i][j] = arr[i][j]
                elif i == 0:
                    dp[i][j] = arr[i][j] + dp[i][j - 1]
                elif j == 0:
                    dp[i][j] = arr[i][j] + dp[i - 1][j]
                else:   
                    dp[i][j] = arr[i][j] + min(dp[i - 1][j], dp[i][j - 1])

        return dp[m - 1][n - 1]
```

```python
grid = [[1,3,1],[1,5,1],[4,2,1]]
obj = Solution()
obj.minPathSum(grid)
```

# `Recursion + Memo` 

```python
from typing import List

class Solution:
    def helper(self, sr, sc, er, ec, grid, dp):
        if sr == er and sc == ec:
            return grid[sr][sc]

        if sr > er or sc > ec:
            return float('inf')

        if dp[sr][sc] != -1:
            return dp[sr][sc]

        right = self.helper(sr, sc + 1, er, ec, grid, dp)
        down = self.helper(sr + 1, sc, er, ec, grid, dp)

        dp[sr][sc] = grid[sr][sc] + min(right, down)

        return dp[sr][sc]

    def minPathSum(self, grid: List[List[int]]) -> int:
        m = len(grid)
        n = len(grid[0])

        dp = [[-1] * n for _ in range(m)]

        return self.helper(0, 0, m - 1, n - 1, grid, dp)
```

```python
grid = [[1,3,1],[1,5,1],[4,2,1]]
obj = Solution()
obj.minPathSum(grid)
```
