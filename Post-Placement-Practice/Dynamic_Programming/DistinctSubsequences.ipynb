# `Recursion`

```python
# TC :- O(Exponential)
# SC :- O(N+M)

def helper(i,j,s,t):
    if j == len(t):
        return 1
    
    if i == len(s):
        return 0
    
    if s[i] == t[j]:
        take = helper(i+1,j+1,s,t)
        notTake = helper(i+1,j,s,t)

        return take + notTake
    else:
        return helper(i+1,j,s,t)

def numDistinct(s, t):
    return helper(0,0,s,t)

s = "babgbag"
t = "bag"

numDistinct(s,t)
```

# `Memoization`

```python
# TC :- O(Exponential)
# SC :- O(N+M) + O(N+M) Stack Space

def helper(i,j,s,t,dp):
    if j == len(t):
        return 1
    
    if i == len(s):
        return 0

    if dp[i][j] != -1:
        return dp[i][j]
    
    if s[i] == t[j]:
        take = helper(i+1,j+1,s,t,dp)
        notTake = helper(i+1,j,s,t,dp)

        dp[i][j] = take + notTake
    else:
        dp[i][j] =  helper(i+1,j,s,t,dp)

    return dp[i][j]

def numDistinct(s, t):
    dp = [[-1] * len(t) for _  in range(len(s))]
    return helper(0,0,s,t,dp)

s = "babgbag"   
t = "bag"

numDistinct(s,t)
```

# `Tabulation`

```python
def numDistinct(s, t):
    n = len(s)
    m = len(t)

    dp = [[0] * (m + 1) for _ in range(n + 1)]

    # Empty t can be formed in exactly 1 way
    for i in range(n + 1):
        dp[i][0] = 1

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if s[i - 1] == t[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + dp[i - 1][j]
            else:
                dp[i][j] = dp[i - 1][j]

    return dp[n][m]

s = "babgbag"
t = "bag"

numDistinct(s,t)
```

# `Space Optimization`

```python
def numDistinct(s, t):
    n = len(s)
    m = len(t)

    prev = [0] * (m + 1)
    curr = [0] * (m + 1)

    # Empty t can be formed in exactly 1 way
    prev[0] = 1
    curr[0] = 1

    for i in range(1, n + 1):

        for j in range(1, m + 1):

            if s[i - 1] == t[j - 1]:
                curr[j] = prev[j - 1] + prev[j]

            else:
                curr[j] = prev[j]

        prev = curr[:]

    return prev[m]


s = "babgbag"
t = "bag"

print(numDistinct(s, t))
```
