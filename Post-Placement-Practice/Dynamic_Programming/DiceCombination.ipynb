```python
MOD = 1000000007

def f(n):
    if n == 0:
        return 1

    total = 0

    for i in range(1, 7):
        if n - i < 0:
            break
        total = (total + f(n - i)) % MOD

    return total


n = int(input())
print(f(n))
```

```python
MOD = 1000000007

def f(n, dp):
    if n == 0:
        return 1

    if dp[n] != -1:
        return dp[n]

    total = 0

    for i in range(1, 7):
        if n - i < 0:
            break
        total = (total + f(n - i, dp)) % MOD

    dp[n] = total
    return dp[n]


n = int(input())
dp = [-1] * (n + 1)
print(f(n, dp))
```
