```python
from sys import flags
def coinscount(i,coins,amount):
    if i == len(coins):
        if amount == 0:
            return 0
        else:
            return float('inf')

    skip = coinscount(i + 1,coins,amount)
    if amount - coins[i] < 0:
        return skip
    
    pick = 1 + coinscount(1,coins,amount - coins[i])
    return min(skip,pick)

def coinchange(coins,amount):
    ans = coinscount(0,coins,amount)
    if ans == float('inf'):
        return -1
    
    return ans

coins = [1,2,5]
amount = 11
coinchange(coins,amount)
```

# `Memoization`

```python
from typing import List
class Solution:
    def coinscount(self,i,coins,amount,dp):
        if i == len(coins):
            if amount == 0:
                return 0
            else:
                return float('inf')

        if dp[i][amount] != -1:
            return dp[i][amount]

        skip = self.coinscount(i + 1,coins,amount,dp)
        
        if amount - coins[i] < 0:
            dp[i][amount] = skip
            return dp[i][amount]
        
        pick = 1 + self.coinscount(i,coins,amount - coins[i],dp)
        dp[i][amount] = min(skip,pick) 
        
        return dp[i][amount]
    
    def coinChange(self, coins: List[int], amount: int) -> int:
        # i and amount is changins so dp will work on that
        dp = [[-1] * (amount + 1) for _ in range(len(coins))]
        ans = self.coinscount(0,coins,amount,dp)
        if ans == float('inf'):
            return -1
        
        return ans
```

```python
coins = [1,2,5]
amount = 11
obj = Solution()
obj.coinChange(coins,amount)
```
