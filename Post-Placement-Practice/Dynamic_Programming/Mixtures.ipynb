```python
from math import inf


def g(colors, i, j):
    result = 0
    for k in range(i, j + 1):
        result = (result + colors[k]) % 100
    return result


def f(colors, i, j):
    if i == j:
        return 0

    result = inf

    for k in range(i, j):
        smoke = (
            f(colors, i, k)
            + f(colors, k + 1, j)
            + g(colors, i, k) * g(colors, k + 1, j)
        )
        result = min(result, smoke)

    return result


while True:
    try:
        n = int(input())
        colors = list(map(int, input().split()))
        print(f(colors, 0, n - 1))
    except EOFError:
        break
```

# `Memoization`

```python
from math import inf


def g(colors, i, j):
    return sum(colors[i:j + 1]) % 100


def f(colors, i, j, dp):
    if i == j:
        return 0

    if dp[i][j] != -1:
        return dp[i][j]

    ans = inf

    for k in range(i, j):
        ans = min(
            ans,
            f(colors, i, k, dp)
            + f(colors, k + 1, j, dp)
            + g(colors, i, k) * g(colors, k + 1, j)
        )

    dp[i][j] = ans
    return ans


while True:
    try:
        n = int(input())
        colors = list(map(int, input().split()))

        dp = [[-1] * n for _ in range(n)]

        print(f(colors, 0, n - 1, dp))

    except EOFError:
        break
```
