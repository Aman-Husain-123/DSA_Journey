```python
# Using DP
from typing import List
class Solution:
    def fibo(self, n: int, dp: List[int]):
        if n <= 1:
            return n

        if dp[n] != -1:
            return dp[n]

        dp[n] = self.fibo(n - 1, dp) + self.fibo(n - 2, dp)
        return dp[n]

    def fib(self, n: int) -> int:
        dp = [-1] * (n + 1)
        return self.fibo(n, dp)
        
```

```python
obj = Solution()
n = 5
dp = [-1] * (n + 1)
obj.fibo(n,dp)
```

# `Tabulation Method (Bottom Up DP)`

```python
def fib(n):
    dp = [0] * (n + 1)
    dp[0] = 0

    if n > 1:
        dp[1] = 1
    
    for i in range(2,n + 1):
        dp[i] = dp[i-1] + dp[i - 2]
    
    return dp[n]
```

```python
fib(5)
```
