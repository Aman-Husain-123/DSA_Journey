```python
def helper(sr,sc,er,ec):
    # Reached destination
    if sr == er and sc == ec:
        return 1
    
    # Out of bounds
    if sc > er or sc > ec:
        return 0
    
    rightways = helper(sr,sc + 1,er,ec)
    downways = helper(sr+1,sc,er,ec)

    return rightways + downways

def uniquepaths(m,n):
    return helper(0,0,m-1,n-1)

m = 3
n = 7
uniquepaths(m,n)
```

```python
# TODO: DP
class Solution:
    def helper(self, sr, sc, er, ec, dp):
        # Reached destination
        if sr == er and sc == ec:
            return 1

        # Out of bounds
        if sr > er or sc > ec:
            return 0

        # Already computed
        if dp[sr][sc] != -1:
            return dp[sr][sc]

        # Move Right + Move Down
        dp[sr][sc] = (
            self.helper(sr, sc + 1, er, ec, dp) +
            self.helper(sr + 1, sc, er, ec, dp)
        )

        return dp[sr][sc]

    def uniquePaths(self, m: int, n: int) -> int:
        dp = [[-1] * n for _ in range(m)]
        return self.helper(0, 0, m - 1, n - 1, dp)

m = 3
n = 7
# uniquepaths(m,n)
obj = Solution()
obj.uniquePaths(m,n)
```

# `Tabulation`

```python
class Solution:
    def uniquePaths(self, m, n):
        # Create DP table
        dp = [[0] * n for _ in range(m)]

        for i in range(m):
            for j in range(n):
                if i == 0 or j == 0:
                    dp[i][j] = 1
                else:
                    dp[i][j] = dp[i - 1][j] + dp[i][j - 1]

        return dp[m - 1][n - 1]


# ---------------- Driver Code ----------------
if __name__ == "__main__":
    m = int(input("Enter number of rows: "))
    n = int(input("Enter number of columns: "))

    obj = Solution()
    print("Unique Paths:", obj.uniquePaths(m, n))
```
