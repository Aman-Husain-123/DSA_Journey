```python
# TLE 
envelopes = [[5,4],[6,4],[6,7],[2,3]]

# Sort by width ↑ and height ↓
envelopes.sort(key=lambda x: (x[0], -x[1]))

# Extract heights
nums = [h for _, h in envelopes]

n = len(nums)
dp = [1] * n

for i in range(1, n):
    for j in range(i):
        if nums[i] > nums[j]:
            if dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1

print("Sorted Envelopes:", envelopes)
print("Heights:", nums)
print("DP:", dp)
print("Answer:", max(dp))
```
