```python
# TC :- 2^n
# TC :- O(n)
def helper(cost,idx):
    if idx == 0 or idx == 1:
        return cost[idx]
    
    return cost[idx] + min(helper(cost,idx - 1),helper(cost,idx - 2))


def minCostClimbingStairs(cost):
    n = len(cost)
    return min(helper(cost,n-1),helper(cost,n-2))
```

# `Top Down DP`

```python
# TC :- 2^n
# TC :- O(n)
def helper(cost,idx,dp):
    if idx == 0 or idx == 1:
        return cost[idx]
    
    if dp[idx] != -1:
        return dp[idx]

    dp[idx] =  cost[idx] + min(helper(cost,idx - 1,dp),helper(cost,idx - 2,dp))

    return dp[idx]

def minCostClimbingStairs(cost):
    n = len(cost)
    dp = [-1] * (n + 1)
    return min(helper(cost,n-1,dp),helper(cost,n-2,dp))
```

```python
cost = [10,15,20]
minCostClimbingStairs(cost)
```

# `Bottom Up DP`

```python
def minCostClimbingStairs(cost):
    n = len(cost)
    dp = [-1] * (n + 1)
    dp[0] = cost[0]
    dp[1] = cost[1]
    for i in range(2,n):
        dp[i] = cost[i] + min(dp[i-1],dp[i-2])
    return min(dp[n-1],dp[n-2])

# Explanation
# cost = [1, 100, 1, 1, 1, 100, 1, 1, 100, 1]

# dp = [1,100,2,3,4,103,5,6,105,101]
```

```python
cost = [1, 100, 1, 1, 1, 100, 1, 1, 100, 1]
minCostClimbingStairs(cost)
```

# `Space Optimization`

```python
# TC :- 2^n
# TC :- O(n)
def helper(cost):
    n = len(cost)
    
    prev1 = 0
    prev2 = 0
    result = prev2

    for i in range(2,n + 1):
        result = min(prev2 + cost[i-1],prev1 + cost[i-2])
        prev1 = prev2
        prev2 = result
    
    return result

def minCostClimbingStairs(cost):
    n = len(cost)
    return min(helper(cost),helper(cost))

cost = [1, 100, 1, 1, 1, 100, 1, 1, 100, 1]
minCostClimbingStairs(cost)
```
