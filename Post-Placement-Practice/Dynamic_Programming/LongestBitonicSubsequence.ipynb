```python
# dp1 :- LIS from front
# dp2 :- LIS from back
# bitonic :- ele = (dp1[i] + dp[j]) - 1
```

```python
class Solution:
    def longestBitonicSequence(self, n, arr):

        # LIS ending at i
        LIS_dp = [1] * n

        for i in range(n):
            for prev in range(i):

                if arr[prev] < arr[i]:
                    LIS_dp[i] = max(
                        LIS_dp[i],
                        LIS_dp[prev] + 1
                    )

        # LDS starting from i
        LDS_dp = [1] * n

        for i in range(n - 1, -1, -1):
            for prev in range(n - 1, i, -1):

                if arr[prev] < arr[i]:
                    LDS_dp[i] = max(
                        LDS_dp[i],
                        LDS_dp[prev] + 1
                    )

        # Find longest bitonic subsequence
        maxLen = 0

        for i in range(n):

            # Peak must have both increasing and decreasing parts
            if LIS_dp[i] > 1 and LDS_dp[i] > 1:
                maxLen = max(
                    maxLen,
                    LIS_dp[i] + LDS_dp[i] - 1
                )

        return maxLen
        
```
