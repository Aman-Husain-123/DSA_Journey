```python
def search_index_position(arr,target):
    n = len(arr)
    low = 0
    high = n - 1
    ans = n

    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] >= target:
            ans = mid
            high = mid - 1
        else:
            low = mid + 1
    
    return ans
```

```python
nums = [1,3,5,6]
target = 5
print(search_index_position(nums,target))
```
