### `TC:` O(min - max + 1) * n

```python
import Binary_Search
from Binary_Search import square
def possible(arr, day, m, k):
    cnt = 0
    noofB = 0

    for i in range(len(arr)):   # corrected
        if arr[i] <= day:
            cnt += 1
        else:
            noofB += (cnt // k)   # corrected
            cnt = 0

    noofB += (cnt // k)   # corrected

    if noofB >= m:
        return True
    else:
        return False


bloomDay = [1,10,3,10,2]
m = 3   # No of bouqets to make
k = 1   # Adjacent no of flowers required

n = len(bloomDay)

if n < m * k:
    print(-1)   # corrected
else:
    minn = min(bloomDay)
    maxx = max(bloomDay)

    for i in range(minn, maxx + 1):   # corrected
        if possible(bloomDay, i, m, k) == True:
            print(i)   # corrected
            break


# TODO: Possible function parameter
# arr	            bloomDay array
# day	            Current day we are checking
# m	                Number of bouquets needed
# k	                Flowers needed per bouquet
```
