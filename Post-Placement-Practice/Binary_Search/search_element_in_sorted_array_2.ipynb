```python
def rotated_sorted_array(arr,target,low,high):

    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] == target:
            return mid
        
        if arr[low] == arr[mid] and arr[mid] == arr[high]:
            low += 1
            high -= 1
            continue
        
        if arr[low] <= arr[mid]:
            if arr[low] <= target and target <= arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] <= target and target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1
```

```python
arr = [4,5,6,7,0,1,2]
target = 0
low = 0
high = len(arr) - 1
print(rotated_sorted_array(arr,target,low,high))
```
