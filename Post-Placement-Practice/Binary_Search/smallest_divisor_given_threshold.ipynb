```python
import math
def sumbyD(arr,div):
    summ = 0
    n = len(arr)
    for i in range(n):
        summ += math.ceil(arr[i] / div)
    return summ
```

```python
# TODO: Binary Search
arr = [44,22,33,11,1]
threshold = 5
low = 1
high = max(arr)

while low <= high:
    mid = low + (high - low) // 2
    if sumbyD(arr,mid) <= threshold:
        high = mid - 1
    else:
        low = mid + 1
print(low)
```
