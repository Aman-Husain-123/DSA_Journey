```python
def upperbound(arr,target):
    low = 0
    n = len(arr)
    high = n - 1
    ans = n

    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] > target:
            ans = mid
            high = mid - 1
        else:
            low = mid + 1
    
    return ans
```

```python
arr = [2, 3, 7, 9,10, 11, 11, 25]
target = 9
print(upperbound(arr,target))
```
