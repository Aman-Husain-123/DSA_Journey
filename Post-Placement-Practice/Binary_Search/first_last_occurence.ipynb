# `O(n) Approach`

```python
nums = [5,7,7,8,8,10]
target = 8
last,first = -1,-1

for i in range(len(nums)):
    if nums[i] == target:
        if first == -1:
            first = i
        last = i

print(first)      
print(last)      
```

# `O(log n) Approach`

```python
def first_last_occurrence(arr, target):
    first, last = -1, -1

    # Find first occurrence
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] == target:
            first = mid
            high = mid - 1   # move left
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    # If not found
    if first == -1:
        return [-1, -1]

    # Find last occurrence
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] == target:
            last = mid
            low = mid + 1    # move right
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return [first, last]


# Test
nums = [5,7,7,8,8,10]
target = 8
print(first_last_occurrence(nums, target))
```
