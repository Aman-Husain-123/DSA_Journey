### `Brtue Approach`

```python
import math

def func(arr, speed):
    summ = 0
    
    for bananas in arr:
        summ += math.ceil(bananas / speed)
    
    return summ


piles = [3, 6, 7, 11]
h = 8

maxx = max(piles)
ans = maxx

for i in range(1, maxx + 1):
    reqtime = func(piles, i)
    
    if reqtime <= h:
        ans = i
        break

print(ans)
```

## `Aoptimal Approach`

```python
import math
def func(arr,speed):
    summ = 0
    for banana in arr:
        summ += math.ceil(banana / speed)
    return summ
```

```python
piles = [3,6,7,11]
h = 8

low = 1
high = max(piles)
ans = high

while low <= high:
    mid = (low + high) // 2
    totalh = func(piles,mid)

    if totalh <= h:
        ans = mid
        high = mid - 1
    else:
        low = mid + 1
print(ans)
```
