```python
def sqrt(x):
    low = 0
    high = x

    while low <= high:
        mid = low + (high - low) // 2
        if mid * mid == x:
            return mid
        elif mid * mid > x:
            high = mid - 1
        else:
            low = mid + 1
    return high
```

```python
x = 8
print(sqrt(x))
```
