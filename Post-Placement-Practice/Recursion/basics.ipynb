```python
def printt(cnt):
    if cnt == 3:
        return     
    print(cnt)
    printt(cnt + 1)  # Pass the incremented value to the next call

printt(0)
```

```python
def f(i,n):
    if i > n:
        return
    
    print("Aman")
    f(i + 1,n)

n = 3
f(1,n)
```

```python
def func(i,n):
    if i < 1:
        return
    
    func(i-1,n)
    print(i)

n = 3
func(3,n)
```

```python
def prod(a,b):
    return a * b

x = prod(2,4)
print(x)
```
