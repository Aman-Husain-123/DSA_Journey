```python
def pow(a,b):
    if a == 0 and b == 0:
        print("0 raised to the power 0 is not defined")
        return -100
    p = 1
    for i in range(1,b + 1):
        p *= a
        print(p)
    
    return p

pow(0,0)
```

```python
# Recursive Solution
def pow(x,n):
    if n == 0:
        return 1
    
    return x * pow(x,n-1)

pow(3,8)
```

```python
# Tatti Way
def pow(x,n):
    if n == 1:
        return x
    
    ans = pow(x,n/2)
    return pow(x,n/2) * pow(x,n/2)

pow(3,4)
```

```python
# Sexy way
def pow(x,n):
    if n == 1:
        return x
    
    ans = pow(x,n/2)
    return ans * ans

pow(3,3)
# Problem in this code 👇
# If number becomes odd so its creating issue
```

```python
def pow(x, n):
    if n == 0:
        return 1

    half = pow(x, n // 2)

    if n % 2 == 0:
        return half * half
    else:
        return half * half * x


print(pow(3, 3))
```
