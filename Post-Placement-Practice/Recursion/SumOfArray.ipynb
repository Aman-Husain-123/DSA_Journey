```python
def f(arr,i):
    # TC : O(n)
    # SC : O(n)
    if i == len(arr):
        return 0
    
    return f(arr,i + 1) + arr[i]L
```

```python
arr = [1,2,3,-1,4,5]
f(arr,0)
```

```python
# f(arr,0)
# │
# └── f(arr,1)
#       │
#       └── f(arr,2)
#             │
#             └── f(arr,3)
#                   │
#                   └── f(arr,4)
#                         │
#                         └── return 0   ← Base Case
```

```python
# Returning Back

# f(arr,4)
# returns 0

# f(arr,3)
# return f(arr,4) + arr[3]
# = 0 + 40
# = 40

# f(arr,2)
# return f(arr,3) + arr[2]
# = 40 + 30
# = 70


# f(arr,1)
# return f(arr,2) + arr[1]
# = 70 + 20
# = 90

# f(arr,0)
# return f(arr,1) + arr[0]
# = 90 + 10
# = 100
```
