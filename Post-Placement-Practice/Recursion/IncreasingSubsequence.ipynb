```python
def printSubset(arr,n,idx,ans,k):
    # Base case
    if idx == n:
        if len(ans) == k:
            print(*ans)
        return
    
    # Dont take the current element
    printSubset(arr,n,idx+1,ans,k)

    # Take the current element
    ans.append(arr[idx])
    printSubset(arr,n,idx+1,ans,k)
    
    # Backtrack
    ans.pop()

arr = [1, 2, 3, 4, 5]
n = len(arr)

k = 3

printSubset(arr, n, 0, [], k)
```
