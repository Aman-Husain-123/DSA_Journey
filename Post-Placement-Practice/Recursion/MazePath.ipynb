```python
def maze(sr,sc,er,ec):
    if sr > er or sc > ec:
        return 0
    
    if sr == er and sc == ec:
        return 1

    rightWays = maze(sr,sc + 1,er,ec)
    leftWays = maze(sr + 1,sc,er,ec)

    return rightWays + leftWays

def printPath(sr,sc,er,ec,s):
    if sr > er or sc > ec:
        return
    
    if sr == er and sc == ec:
        print(s)
        return

    printPath(sr,sc + 1,er,ec,s+"R") # Right
    printPath(sr + 1,sc,er,ec,s+"D") # Down

    

printPath(1,1,3,3,"")
```

# `Using 2 Variable`

```python
#               Column
#            1       2       3
#         ┌───────┬───────┬───────┐
# Row 1   │ (1,1) │ (1,2) │ (1,3) │
#         ├───────┼───────┼───────┤
# Row 2   │ (2,1) │ (2,2) │ (2,3) │
#         ├───────┼───────┼───────┤
# Row 3   │ (3,1) │ (3,2) │ (3,3) │
#         └───────┴───────┴───────┘
```

```python
def maze(row,col):
    if row == 1 or col == 1:
        return 1
    
    rightways = maze(row,col-1)
    downways = maze(row-1,col)

    return rightways + downways

def printPath(row,col,s):
    if row == 1 and col == 1:
        print(s)
        return
    
    if col > 1:
        printPath(row, col - 1, s + "R")

    if row > 1:
        printPath(row - 1, col, s + "D")


print("Total paths:", maze(3, 3))
print("Paths:")
printPath(3, 3, "")
```
