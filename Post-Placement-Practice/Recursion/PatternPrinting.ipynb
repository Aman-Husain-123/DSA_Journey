```python
def f(n,r):
    if r > n:
        return
    
    for i in range(1,r+1):
        print("*", end="")
    print()

    f(n,r+1)

n = 5
r = 1
f(n,r)
```

```python
def f(n,r,c):
    if r > n:
        return
    
    if c > r:
        # Current row is done, move to next row
        print()
        f(n,r+1,1)
        return
    print("*", end="")
    f(n,r,c+1)
    
n = 8
f(n,1,1)
```

```python
#                  r       c
#                  ↓       ↓

# f(8,1,1)       row 1    star 1
#      ↓
# f(8,1,2)       row 1    done
#      ↓
# f(8,2,1)       row 2    star 1
#      ↓
# f(8,2,2)       row 2    star 2
#      ↓
# f(8,2,3)       row 2    done
#      ↓
# f(8,3,1)       row 3    star 1
#      ↓
# f(8,3,2)       row 3    star 2
#      ↓
# f(8,3,3)       row 3    star 3
#      ↓
# f(8,3,4)       row 3    done
#      ↓
# f(8,4,1)       row 4    star 1
#      ↓
# ...
```
