```python
def permutations(ans, original):

    if original == "":
        print(ans)
        return

    for i in range(len(original)):

        ch = original[i]    # 1.a  2.b  3.c

        left = original[:i]
        right = original[i + 1:]

        permutations(ans + ch, left + right)


s = "abc"

permutations("", s)
```
