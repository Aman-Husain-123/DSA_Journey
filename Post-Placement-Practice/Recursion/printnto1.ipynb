```python
def func(n):
    if n == 0:
        return

    print(n)
    func(n-1)

func(4)
```

```python
# print1 - n
def func(n):
    if n == 0:
        return

    func(n-1)
    print(n)

func(4)
```

```python
#             CALLING ↓              RETURNING ↑

# func(4) ─────────────────────────── print(4)
#    ↓                                   ↑
# func(3) ─────────────────────────── print(3)
#    ↓                                   ↑
# func(2) ─────────────────────────── print(2)
#    ↓                                   ↑
# func(1) ─────────────────────────── print(1)
#    ↓                                   ↑
# func(0) ─────── return ────────────────┘
#                  🛑
```

```python
# TODO: 2 Variable
def func(i:int,n:int) -> int:
    if i > n:
        return
    
    print(i)
    func(i+1,n)

func(1,4)
```

```python
# TODO: print sum from 1 - n (Parameterized)

def func(summ:int,n:int) -> int:
    if n == 0:
        print(summ)
        return

    func(summ + n,n-1)
    
func(0,6)
```

```python
def func(n):
    if n == 0:
        return 0
    
    summ = 0
    
    summ = n + func(n-1)

    return summ


func(6)
```
