```python
def generate(s,n):
    if n == len(s):
        print(s)
        return
    
    generate(s + '0', n)    
    generate(s + '1', n)

n = 3
generate('', n)
```

```python
# No Consecutive 1's
def generate(s,n):
    if n == len(s):
        print(s)
        return
    
    generate(s + '0', n)
    
    if s == '' or s[-1] != '1':
        generate(s + '1', n)


n = 3
generate('', n)
```
