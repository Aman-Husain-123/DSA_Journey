```python
# Mention the question?
# Combination sum problem

def combinationSum(v,arr,n,target):
    if target == 0:
        for i in range(len(v)):
            print(v[i], end=" ")
        print()
    
    if target < 0:
        return
    
    for i in range(n):
        v.append(arr[i])
        combinationSum(v, arr, n, target - arr[i])
        v.pop()
```

```python
arr = [2,3,5]
n = len(arr)
v = []
combinationSum(v,arr,n,8)
```

```python
# TODO: Only distinct combinations are allowed. So, we need to modify the code to avoid duplicates. We can do this by adding an index parameter to the function and only allowing elements from that index onwards to be added to the combination. This way, we avoid using the same element multiple times in different orders.
def combinationSum(v,arr,n,target,index):
    if target == 0:
        for i in range(len(v)):
            print(v[i], end=" ")
        print()

    if target < 0:
        return

    for i in range(index, n):
        v.append(arr[i])
        combinationSum(v, arr, n, target - arr[i], i)
        v.pop()

arr = [2,3,5]
n = len(arr)
v = []
combinationSum(v,arr,n,8,0)
```
