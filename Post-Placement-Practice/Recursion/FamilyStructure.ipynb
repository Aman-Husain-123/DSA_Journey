```python
def f(root, n, k):

    if n == 1:
        return root

    value = (1 << (n - 1)) // 2

    if k <= value:
        # Jump left
        return f(root, n - 1, k)

    else:
        new_root = 'F' if root == 'M' else 'M'

        return f(new_root, n - 1, k - value)


# Number of test cases
t = int(input())

while t > 0:

    n, k = map(int, input().split())

    result = f('M', n, k)

    if result == 'M':
        print("Male")
    else:
        print("Female")

    t -= 1
```
