```python
def removechar(ans,original):
    if len(original) == 0:
        print(ans,end="")
        return
    
    ch = original[0]

    if ch == 'a':
        removechar(ans,original[1:])
    else:
        removechar(ans + ch,original[1:])

s = "baccad"
removechar("",s)
```

```python
def removechar(ans,original,idx):
    if len(original) == 0:
        print(ans,end="")
        return
    
    ch = original[idx]

    if ch == 'a':
        removechar(ans,original,idx+1)
    else:
        removechar(ans + ch,original,idx+1)

s = "baccad"
removechar("",s,0)
```
