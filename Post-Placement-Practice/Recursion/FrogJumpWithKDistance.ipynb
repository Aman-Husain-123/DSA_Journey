```python
class Solution:
    def solve(self, i, height, k):
        n = len(height)

        # Base case: reached last stone
        if i == n - 1:
            return 0

        ans = float('inf')

        # Try all jumps from 1 to K
        for j in range(1, k + 1):

            # Cannot jump outside the array
            if i + j >= n:
                break

            cost = (
                self.solve(i + j, height, k)
                + abs(height[i] - height[i + j])
            )

            ans = min(ans, cost)

        return ans

    def minCost(self, height, k):
        return self.solve(0, height, k)

```
