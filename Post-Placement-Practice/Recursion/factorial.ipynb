```python
def factUpto(n):
    for j in range(1, n + 1):
        f = 1

        for i in range(2, j + 1):
            f *= i

        print(f)


factUpto(5)
```

```python
# TODO: So base case needs to be introduced
def fact(n:int) -> int:
    if n == 1 or n == 0:
        return 1
    
    return n * fact(n-1)


fact(0)

# Tree 👇
```
