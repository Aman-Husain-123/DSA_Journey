```python
def f(arr,n,idx,maxx):
    if idx == n:
        print(maxx)
        return
    if maxx < arr[idx]:
        maxx = arr[idx]
    
    f(arr,n,idx+1,maxx)
```

```python
arr = [32,21,4,323,2]
n = len(arr)
f(arr, n,0,float('-inf'))
```
