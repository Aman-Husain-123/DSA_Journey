```python
class Solution:
    def __init__(self):
        self.mapping = [
            "",      # 0
            "",      # 1
            "abc",   # 2
            "def",   # 3
            "ghi",   # 4
            "jkl",   # 5
            "mno",   # 6
            "pqrs",  # 7
            "tuv",   # 8
            "wxyz"   # 9
        ]
        self.result = []

    def f(self, digits, i, output):
        # Base Case
        if i == len(digits):
            self.result.append(output)
            return

        # Current digit
        digit = int(digits[i])

        # Characters corresponding to the digit
        chars = self.mapping[digit]

        # Try every character
        for ch in chars:
            self.f(digits, i + 1, output + ch)

    def letterCombinations(self, digits):
        if len(digits) == 0:
            return []

        self.f(digits, 0, "")
        return self.result


# Driver Code
obj = Solution()
print(obj.letterCombinations("23"))
```
