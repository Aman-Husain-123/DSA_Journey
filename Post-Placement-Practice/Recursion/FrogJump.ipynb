```python
class Solution:
    def solve(self, ind, height):
        # Base case
        if ind == 0:
            return 0
        
        # if dp[ind] != -1:
        #     return dp[ind]

        jumpOne = self.solve(ind - 1, height) + abs(height[ind] - height[ind - 1])

        jumpTwo = float('inf')
        if ind > 1:
            jumpTwo = self.solve(ind - 2, height) + abs(height[ind] - height[ind - 2])

        
        return min(jumpOne, jumpTwo)
        
    def minCost(self, height):
        if not height:
            return 0
        

        n = len(height)
        # dp = [-1]  * n
        return self.solve(n - 1, height)

height = [10,30,40,20]
obj = Solution()
obj.minCost(height)
```
