```python
class Solution:
    def f(self,n):
        if n == 0:
            return 0
        
        if n == 1:
            return 1
        
        if n == 2:
            return 2
        
        return self.f(n-1) + self.f(n-2)
        
    def numberOfWays(self, n):
        # code here
        return self.f(n)

Obj = Solution()
obj.numberOfWays(4)
```
