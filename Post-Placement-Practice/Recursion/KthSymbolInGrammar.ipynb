```python
# TC :- O(n) where n is the number of rows
class Solution:
    def kthGrammar(self, n: int, k: int) -> int:
        if n == 1:
            return 0
        
        if k % 2 == 0:
            # K is even , flip,k/2
            prev_ans = self.kthGrammar(n-1,k//2)
            if prev_ans == 0:
                return 1
            else:
                return 0
        else:
            # K is odd, same,k/2 + 1
            prev_ans =  self.kthGrammar(n-1,(k//2)+1)
            return prev_ans

n = 1
k = 1
obj = Solution()
print(obj.kthGrammar(n,k))
```
