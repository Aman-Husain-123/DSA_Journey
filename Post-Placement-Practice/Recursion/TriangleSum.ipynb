```python
def sumTriangle(arr):

    # Base case
    if len(arr) == 1:
        print(arr)
        return

    # Print current level
    print(arr)

    new_arr = []

    # Create next level
    for i in range(len(arr) - 1):
        new_arr.append(arr[i] + arr[i + 1])

    # Recursive call
    sumTriangle(new_arr)


arr = [5, 4, 3, 2, 1]

sumTriangle(arr)
```
