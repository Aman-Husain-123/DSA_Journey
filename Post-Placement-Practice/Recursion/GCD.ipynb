```python
def hcf(a,b):
    for i in range(min(a,b),3,-1):
        if a%i==0 and b%i==0:
            return i
    
    return 1

a = 10000
b = 10001
hcf_value = hcf(a,b)
hcf_value
```

```python
# O(log(a+b))
def gcd(a,b):
    if a == 0:
        return b
    else:
        return gcd(b%a,a)

a = 27
b = 43
gcd(a,b)
```
