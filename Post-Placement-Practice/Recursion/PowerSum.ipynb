```python
# Problem Link👇
# https://www.hackerrank.com/challenges/the-power-sum/problem

def solve(remaining,num):
    # Found one valid combination
    if remaining == 0:
        return 1
    
    # Invalid combination
    if remaining < 0:
        return 0
    
    # If num^n is greater than remaining, we cannot include it in the combination
    power = num ** n

    # No future numbers can be included if power is greater than remaining
    if power > remaining:
        return 0
    
    # Use the current number and move to the next number
    take = solve(remaining - power, num + 1)

    # Dont use the current number and move to the next number
    dont_take = solve(remaining, num + 1)

    return take + dont_take

# driver code
if __name__ == "__main__":
    x = int(input())
    n = int(input())
    result = solve(x, 1)
    print(result)
```
