```python
def f(string, i, out):

    if i == len(string):
        print(out)
        return

    if i == len(string) - 1:
        # We are at the last index
        one = int(string[i])

        f(string, i + 1, out + chr(64 + one))

    else:
        one = int(string[i])

        # Take one digit
        f(string, i + 1, out + chr(64 + one))

        # Take two digits
        c1 = int(string[i])
        c2 = int(string[i + 1])

        two = c1 * 10 + c2

        if two <= 26:
            f(string, i + 2, out + chr(64 + two))


string = "123"
f(string, 0, "")
```
