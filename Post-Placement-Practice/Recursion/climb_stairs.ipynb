```python
def climb(n):
    if n == 2:
        return 2

    if n == 1:
        return 1

    if n == 3:
        return 3

    return climb(n - 1) + climb(n - 2) + climb(n - 3)
    

print(climb(3))
```

```python
# For 1 and 2 step jump
def climb(n):
    if n == 2:
        return 2

    if n == 1:
        return 1

    return climb(n-1) + climb(n-2)

climb(3) 
```
