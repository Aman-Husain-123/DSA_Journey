```python
def f(ans,original):
    if original == "":
        print(ans)
        return
    
    ch = original[0]
    f(ans,original[1:])
    f(ans+ch,original[1:])

s = "abc"
f("",s)
```

```python
# Using index
def f(ans,original,idx):
    if idx == len(original):
        print(ans)
        return
    
    ch = original[idx]
    f(ans,original,idx+1)
    f(ans+ch,original,idx+1)

s = "abc"
f("",s,0)
```

```python
# Now on int element arr
def f(ans,original,idx):
    if idx == len(original):
        print(ans)
        return
    
    ch = original[idx]
    f(ans,original,idx+1)
    f(ans+[ch],original,idx+1)

s = [1,2,3]
f([],s,0)
```

```python
# Return ans
def f(ans,original,idx):
    if idx == len(original):
        return [ans]
    
    ch = original[idx]
    left = f(ans,original,idx+1)
    right = f(ans+[ch],original,idx+1)
    
    return left + right

s = [1,2,3]
f([],s,0)
```
