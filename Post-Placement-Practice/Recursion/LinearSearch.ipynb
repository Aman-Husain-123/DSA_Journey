```python
def f(arr, i, x):
    # Base case: reached end of array
    if i == len(arr):
        return False

    return (arr[i] == x) or f(arr, i + 1, x)


arr = [32, 324, 322, 12, 4]
x = 12

print(f(arr, 0, x))
```

```python
def LinearSearch(arr,target,n,i):
    if i == n:
        return -1
    if arr[i] == target:
        return i
    return LinearSearch(arr,target,n,i+1)

arr = [1,2,3,4,5]
target = 4
n = len(arr)
LinearSearch(arr,target,n,0)
```
