```python
def f(strr,i,output):
    if i == len(strr) - 1:
        print(output)
        return
    
    if strr[i] == 'x':
        f(strr,i+1,output)
    else:
        f(strr,i+1,output+strr[i])
    
s = "maabxxdx"
f(s,0,"")

# TC :- O(n^2)
# SC :- O(n)
```

```python
# Optimized Code

def f(strr,i,output):
    if i == len(strr) - 1:
        return
    print("".join(output))
    
    if strr[i] == 'x':
        f(strr,i+1,output)
    else:
        output.append(strr[i])
        f(strr,i+1,output)
    
s = "maabxxdx"
output = []
f(s,0,output)

# TC :- O(n^2)
# SC :- O(n)
```
