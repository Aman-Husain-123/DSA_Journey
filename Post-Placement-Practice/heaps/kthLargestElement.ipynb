```python
import heapq
import heapq
from typing import List
def brute_approach(nums:List,k:int) -> List:
    ans = []
    n = len(nums)

    for i in range(k):
        heapq.heappush(ans,nums[i])
    
    for i in range(k,n):
        if nums[i] > ans[0]:
            heapq.heappop(ans)
            heapq.heappush(ans,nums[i])
    
    return ans[0]
```

```python
nums = [3,2,1,5,6,4]
k = 2
brute_approach(nums,k)
```
