```python
class Solution:
    def heapifyDown(self,arr,ind,n):
        smallest = ind
        left = 2 * ind + 1
        right = 2 * ind + 2

        if left < n and arr[left] < arr[smallest]:
            smallest = left

        if right < n and arr[right] < arr[smallest]:
            smallest = right

        if smallest != ind:
            arr[ind], arr[smallest] = arr[smallest], arr[ind]
            self.heapifyDown(arr, smallest, n)
    
    def buildHeap(self, arr):
        n = len(arr)

        # Start from the last non-leaf node
        for i in range((n // 2) - 1, -1, -1):   
            self.heapifyDown(arr, i, n)

        return arr
```

```python
def main():
    obj = Solution()
    arr = [10, 5, 20, 2, 4, 8, 15]

    print("Original Array")
    print(arr)

    obj.buildHeap(arr)

    print("Min heap:")
    print(arr)

if __name__ == "__main__":
    main()
```
