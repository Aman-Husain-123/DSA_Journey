```python
import heapq
class MinHeap:
    def __init__(self):
        self.heap = []
    
    def insert(self,val):
        self.heap.append(val)
        self.heapify_Up(len(self.heap) - 1)
    
    def heapify_Up(self,index):
        while index > 0:
            parent = parent = (index - 1) // 2

            if self.heap[parent] <= self.heap[index]:
                break

            self.heap[parent], self.heap[index] = (
                self.heap[index],
                self.heap[parent],
            )

            index = parent
```
