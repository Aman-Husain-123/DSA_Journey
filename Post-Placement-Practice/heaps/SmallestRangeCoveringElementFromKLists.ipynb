```python
import heapq

nums = [
    [4, 10, 15, 24, 26],
    [0, 9, 12, 20],
    [5, 18, 22, 30]
]

heap = []
current_max = float('-inf')

# Insert first element from every list
for row in range(len(nums)):
    value = nums[row][0]
    heapq.heappush(heap, (value, row, 0))
    current_max = max(current_max, value)

while True:

    # Current minimum
    smallest, row, col = heap[0]

    # Current range
    smallest_range = current_max - smallest

    print("--------------------------------")
    print("Heap :", heap)
    print("Min :", smallest)
    print("Max :", current_max)
    print("Range :", smallest_range)

    # Remove smallest element
    smallest, row, col = heapq.heappop(heap)

    # If this list is exhausted, stop
    if col + 1 == len(nums[row]):
        break

    # Take the next element from the same list
    next_value = nums[row][col + 1]

    heapq.heappush(heap, (next_value, row, col + 1))

    current_max = max(current_max, next_value)
```

```python
# Algorithm

# 1. Create a min heap that stores a tuple of (value, row, col).
#    - value → Current element
#    - row   → Index of the list to which the element belongs
#    - col   → Index of the element within that list

# 2. Initially, insert the first (smallest) element from every sorted list into the heap along with its row and column indices.

# 3. While inserting these elements, maintain a variable current_max, which stores the maximum element currently present in the heap.

# 4. Since it is a min heap, the top element (heap[0]) will always be the smallest among all the current elements.

# 5. Compute the initial range as:
#       current_range = current_max - heap[0].value
#    Store this as the best range initially.

# 6. Repeat the following steps until one of the lists gets exhausted:

#    a. Remove (pop) the smallest element from the heap.

#    b. Compare the current range with the best range found so far.
#       If the current range is smaller, update the answer.

#    c. Move only the list from which the smallest element was removed.

#    d. Check whether the next element exists in the same list.
#       - If it exists, insert it into the heap along with its updated row and column indices.
#       - If it does not exist, terminate the algorithm because a valid range containing one element from every list can no longer be formed.

#    e. Update current_max if the newly inserted element is greater than the existing current_max.

# 7. Continue repeating the above steps until any one of the lists becomes exhausted.

# 8. Return the smallest range obtained during the entire process.

# ------------------------------------------------------------

# Greedy Intuition

# • The heap always contains exactly one element from each list.

# • The smallest element in the heap determines the left boundary of the current range.

# • The largest element (current_max) determines the right boundary of the current range.

# • To reduce the range, we should only move the current minimum element forward.

# • Moving any other element either keeps the minimum unchanged or increases the maximum, so it cannot produce a smaller range.

# Therefore, the greedy choice is to always advance the list containing the current minimum element.

# ------------------------------------------------------------

# Time Complexity:
# O(N log K)

# where:
# N = Total number of elements across all lists
# K = Number of lists

# Space Complexity:
# O(K)
```
