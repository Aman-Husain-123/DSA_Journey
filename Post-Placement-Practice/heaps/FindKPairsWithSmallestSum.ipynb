```python
nums1 = [1, 7, 11]
nums2 = [2, 4, 6]
k = 3

ans = []
min_sum = float('inf')

for i in range(len(nums1)):
    for j in range(len(nums2)):

        curr_sum = nums1[i] + nums2[j]

        if curr_sum < min_sum:
            min_sum = curr_sum
            ans = [[nums1[i], nums2[j]]]

        elif curr_sum == min_sum:
            ans.append([nums1[i], nums2[j]])

print(ans)
```
