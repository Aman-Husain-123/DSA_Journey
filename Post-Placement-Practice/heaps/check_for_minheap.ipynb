```python
class Solution:
    def init(self):
        pass

    def isHeap(self,arr):
        n = len(arr)-1
        for ind in range((n//2) -1,-1,-1):
            l_child = (2 * ind) + 1
            r_child = (2 * ind) + 2

            if l_child < n and arr[l_child] < arr[ind]:
                return False
            if l_child < n and arr[r_child] < arr[ind]:
                return False
        return True
```

```python
def main():
    obj = Solution()

    test_cases = [
        [10, 20, 30, 25, 40, 35],     # True
        [10, 15, 14, 25, 30, 16],     # True
        [20, 10, 30, 40, 50],         # False
        [5],                          # True
        [],                           # True (empty heap)
        [1, 2, 3, 4, 5, 6, 7],        # True
        [1, 3, 2, 7, 6, 4, 5],        # True
        [1, 4, 2, 3],                 # False
    ]

    for i, arr in enumerate(test_cases, 1):
        print(f"Test Case {i}: {arr}")
        if obj.isHeap(arr):
            print("Output : True")
        else:
            print("Output : False")
        print("-" * 30)

if __name__ == "__main__":
    main()
```
