```python
import heapq

matrix = [[1,5,9],[10,11,13],[12,13,15]]

min_heap = []
ans = []
k = 8

# Push every element
for row in matrix:
    for num in row:
        heapq.heappush(min_heap, num)

# Pop in sorted order
while min_heap:
    ans.append(heapq.heappop(min_heap))
ans
```

```python
temp = []
for num in ans:
    heapq.heappush(temp,num)
    if len(temp) > k:
        heapq.heappop(temp)

print(temp[0])
```
