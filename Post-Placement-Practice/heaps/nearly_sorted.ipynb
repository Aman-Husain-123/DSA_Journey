```python
import heapq

arr = [10, 5, 30, 17, 7, 40, 50]
k = 4

pq = []
idx = 0

for i in range(len(arr)):
    heapq.heappush(pq, arr[i])
    

    if len(pq) > k:
        arr[idx] = heapq.heappop(pq)
        print(arr[idx])
        idx += 1
    print(pq)

# Put the remaining elements back into arr
while pq:
    arr[idx] = heapq.heappop(pq)
    idx += 1

print(arr)
```
