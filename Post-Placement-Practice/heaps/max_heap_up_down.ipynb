```python
class MaxHeap:
    def __init__(self):
        self.heap = []

    # -------------------------
    # Heapify Up
    # -------------------------
    def heapify_up(self, index):
        while index > 0:
            parent = (index - 1) // 2

            if self.heap[parent] < self.heap[index]:
                self.heap[parent], self.heap[index] = (
                    self.heap[index],
                    self.heap[parent],
                )
                index = parent
            else:
                break

    # -------------------------
    # Heapify Down
    # -------------------------
    def heapify_down(self, index):
        n = len(self.heap)

        while True:
            largest = index
            left = 2 * index + 1
            right = 2 * index + 2

            if left < n and self.heap[left] > self.heap[largest]:
                largest = left

            if right < n and self.heap[right] > self.heap[largest]:
                largest = right

            if largest == index:
                break

            self.heap[index], self.heap[largest] = (
                self.heap[largest],
                self.heap[index],
            )

            index = largest

    # -------------------------
    # Insert
    # -------------------------
    def insert(self, value):
        self.heap.append(value)
        self.heapify_up(len(self.heap) - 1)

    # -------------------------
    # Delete Maximum
    # -------------------------
    def delete_max(self):
        if not self.heap:
            return None

        if len(self.heap) == 1:
            return self.heap.pop()

        maximum = self.heap[0]

        self.heap[0] = self.heap.pop()
        self.heapify_down(0)

        return maximum


# Example
h = MaxHeap()

nums = [40, 10, 30, 50, 60, 15]

for num in nums:
    h.insert(num)

print("Heap:", h.heap)

print("Deleted:", h.delete_max())
print("Heap after deletion:", h.heap)
```
