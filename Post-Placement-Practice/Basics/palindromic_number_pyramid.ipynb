```python
n = 10
spaces = 2 * (n - 1)
for i in range(1,n + 1):

    # Number
    for j in range(1,i+1):
        print(j,end = " ")
    
    # Spaces
    for k in range(1,spaces + 1):
        print(" ",end = " ")

    # Number
    for j in range(i,0,-1):
        print(j,end = " ")


    print()
    spaces -= 2
 
```
