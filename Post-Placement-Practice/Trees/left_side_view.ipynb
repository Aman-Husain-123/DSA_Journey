```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
```

```python
A = TreeNode("A")
B = TreeNode("B")
C = TreeNode("C")
D = TreeNode("D")
E = TreeNode("E")
F = TreeNode("F")
```

```python
A.left = B
A.right = C
B.left = D
B.right = E
D.left = F
```

```python
from collections import deque

def left_side_view(root):
    if not root:
        return []
    
    q = deque([root])
    ans = []

    while q:
        level_size = len(q)

        for i in range(level_size):
            node = q.popleft()

            if i == 0:
                ans.append(node.val)
            
            if node.left:
                q.append(node.left)
            
            if node.right:
                q.append(node.right)

    return ans
    
    
```

```python
left_side_view(A)
```
