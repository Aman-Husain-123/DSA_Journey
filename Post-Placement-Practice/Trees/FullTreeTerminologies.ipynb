```python
# Program to print sum of all the elements of a binary tree

class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def sumBT(self, root):
        
        if root is None:
            return 0

        return (root.data + self.sumBT(root.left) + self.sumBT(root.right))
    
    def size(self,root):
        if root is None:
            return 0
        
        left = self.size(root.left)
        right = self.size(root.right)

        return left + right + 1
    
    def max(self,root):
        if root is None:
            return float('-inf')
        
        left = self.max(root.left)
        right = self.max(root.right)

        return max(root.data,left,right)

    def min(self,root):
        if root is None:
            return float('inf')
        
        left = self.min(root.left)
        right = self.min(root.right)

        return min(root.data,left,right)
    
    def product(self,root):
        if root is None:
            return 1
        
        left = self.product(root.left)
        right = self.product(root.right)

        return left * right * root.data

# Driver code
def main():
    root = Node(15)
    root.left = Node(10)
    root.right = Node(20)
    root.left.left = Node(8)
    root.left.right = Node(12)
    root.right.left = Node(16)
    root.right.right = Node(25)

    obj = Solution()
    sum_val = obj.sumBT(root)
    size = obj.size(root)
    maxx = obj.max(root)
    minn = obj.min(root)
    # levels = obj.level(root)
    product = obj.product(root)

    print("Sum of all the elements is:", sum_val)
    print("Size of the tree:", size)
    print("Maximum of tree:",maxx)
    print("Minimum of tree:",minn)
    # print("Levels of tree:",levels)
    print("Product of tree nodees:",product)


if __name__ == "__main__":
    main()
```
