```python
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None
```

```python
class Solution:
    def getPath(self,root,arr,x):
        # Base Case
        if root is None:
            return False
        
        # Add current node to path
        arr.append(root.val)

        # If target node is found
        if root.val == x:
            return True
        
        # Search in left or right subtree
        if (self.getPath(root.left,arr,x) or (self.getPath(root.right,arr,x))):
            return True
        
        # Backtracking
        arr.pop()

        return False
    
    def solve(self,root,target):
        arr = []
        if root is None:
            return arr
        
        self.getPath(root,arr,target)
        return arr

# ---------------- DRIVER CODE ----------------

# Creating binary tree
#
#          1
#        /   \
#       2     3
#      / \   / \
#     4   5 6   7
#

root = TreeNode(1)

root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

root.right.left = TreeNode(6)
root.right.right = TreeNode(7)

# Target node
target = 5

# Object of Solution class
obj = Solution()

# Get path
result = obj.solve(root, target)

# Print result
print("Path from root to node:", result)
```
