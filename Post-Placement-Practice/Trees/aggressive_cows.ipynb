```python
def ditributeCows(arr,n,c) -> int:
    arr.sort()
    low = 0
    maxx = 0
    for i in range(len(arr)):
        maxx = max(maxx,arr[i])
    
    high = maxx
    ans = 0
    while low <= high:
        # Mid
        mid = (low + high) // 2
        if good(arr,mid,n,c):
            ans = mid
            low = high + 1
        else:
            high = mid - 1
    return ans
```

```python
def good(arr,mid,n,c):
    # Objective :- Decide whther cows can be placed atleast mid distance apart
    count = 1
    last_position = arr[0]

    for i in range(len(arr)):
        if arr[i] - last_position >= mid:
            last_position = arr[i]
            count += 1
        
        if count == c:
            return True
        
    return False 
```

```python
n = 5
c = 3

arr = [1,2,8,4,9]
arr.sort()
print(ditributeCows(arr,n,c))
```
