```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
```

```python
A = TreeNode('A')
B = TreeNode('B')
C = TreeNode('C')
D = TreeNode('D')
E = TreeNode('E')
F = TreeNode('F')
G = TreeNode('G')
```

```python
A.left = B
A.right = C
B.left = D
B.right = E
C.right = F
F.right = G
```

```python
from collections import deque  # ❌ was: dequeue (wrong name), and wrong import style

def bfs(root):
    node = deque([root])
    res = []

    while node:
        size = len(node)
        level = []

        for _ in range(size):
            curr = node.popleft()  # ❌ was: node = node.popleft() (overwrites the deque!)
            level.append(curr.val)  # fixed accordingly

            if curr.left:
                node.append(curr.left)   # ❌ was: node.append(node.left) (node is a deque, not a tree node)
            if curr.right:
                node.append(curr.right)  # ❌ was: node.append(node.left) (copy-paste bug, should be .right)

        res.append(level)  # ✅ this was outside the for loop, which is correct
    return res
```
