```python
# TreeNode structure
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

# Solution class
class Solution:
    def buildTree(self, preorder, inorder):

        # value -> index mapping of inorder traversal
        self.in_map = {}

        for i in range(len(inorder)):
            self.in_map[inorder[i]] = i

        # Global preorder index
        self.idx = 0

        return self.fun(preorder, 0, len(inorder) - 1)

    def fun(self, preorder, low, high):

        # ---------------- BASE CASE ----------------
        if low > high:
            return None
        # -------------------------------------------

        node = TreeNode(preorder[self.idx])
        self.idx += 1

        id = self.in_map[node.val]

        # Left Subtree
        node.left = self.fun(preorder, low, id - 1)

        # Right Subtree
        node.right = self.fun(preorder, id + 1, high)

        return node

# Inorder print
def printInorder(root):
    if not root:
        return
    printInorder(root.left)
    print(root.val, end=" ")
    printInorder(root.right)

# Driver
if __name__ == "__main__":
    inorder = [9, 3, 15, 20, 7]
    preorder = [3, 9, 20, 15, 7]

    sol = Solution()
    root = sol.buildTree(preorder, inorder)

    print("Inorder of Unique Binary Tree Created:")
    printInorder(root)
    print()
```
