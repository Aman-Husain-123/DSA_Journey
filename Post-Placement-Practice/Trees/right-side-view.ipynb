```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
```

```python
A = TreeNode('A')
B = TreeNode('B')
C = TreeNode('C')
D = TreeNode('D')
E = TreeNode('E')
F = TreeNode('F')


# Edges Connection
A.left = B
A.right = C
B.left = F
C.right = D
D.left = E
```

```python
def rightSideView(root):
    ds = []  # shared result list, persists across recursive calls

    def helper(node, level):
        if node is None:
            return

        # First node we encounter at this depth is the rightmost
        if level == len(ds):
            ds.append(node.val)

        # Visit right subtree first so rightmost node is seen first
        helper(node.right, level + 1)
        helper(node.left, level + 1)

    helper(root, 0)
    return ds
```

```python
rightSideView(A)
```
