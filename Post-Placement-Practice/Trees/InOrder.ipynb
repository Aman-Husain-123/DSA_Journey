# `Recursive Solution`

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def inorder(root):
    if root is None:
        return []

    # Left subtree
    left = inorder(root.left)

    # Root
    result = [root.data]

    # Right subtree
    right = inorder(root.right)

    # Combine: Left -> Root -> Right
    return left + result + right


# Driver Code
if __name__ == "__main__":

    #        1
    #       / \
    #      2   3
    #     / \   \
    #    4   5   6

    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)

    root.left.left = Node(4)
    root.left.right = Node(5)

    root.right.right = Node(6)

    result = inorder(root)

    print(" -> ".join(map(str, result)))
```

# `Iterative Solution`

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None
```

```python
def InorderTraversal(root):
    if not root:
        return []
    
    result = []
    stack = []
    curr = root

    while curr is not None or stack:
        while curr is not None:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop()
        result.append(curr.val)
        curr = curr.right
    
    return result

# Driver Code
if __name__ == "__main__":

    #        1
    #       / \
    #      2   3
    #     / \   \
    #    4   5   6

    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)

    root.left.left = Node(4)
    root.left.right = Node(5)

    root.right.right = Node(6)

    result = inorder(root)

    print(" -> ".join(map(str, result)))
```
