```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None


        #               8
        #              / \\
        #             5   11
        #            / \\
        #           3   7
        #                \\
        #                 10    

root = TreeNode(6)
root.left = TreeNode(4)
root.right = TreeNode(8)
root.left.left = TreeNode(1)
root.left.right = TreeNode(5)
root.left.right.right = TreeNode(7)
```

```python
def righmostvalinleftsubtree(root):
    ans = None
    while root is not None:
        ans = root
        root = root.right
    
    return ans

def leftmostvalueinrightsubtree(root):
    ans = None
    while root is not None:
        ans = root
        root = root.left
    return ans




def getPredSucc(root,key):
    curr = root
    pred = None
    succ = None

    while curr != None:
        if key < curr.val:
            succ = curr
            curr = curr.left
        elif key > curr.val:
            pred = curr
            curr = curr.right
        else:
            if curr.left != None:
                # Inorder pred
                pred = righmostvalinleftsubtree(curr.left)
            
            if curr.right != None:
                # Inorder succ
                succ = leftmostvalueinrightsubtree(curr.right)
            break
    
    pred_val = pred.val if pred else None
    succ_val = succ.val if succ else None
    return pred_val, succ_val
```
