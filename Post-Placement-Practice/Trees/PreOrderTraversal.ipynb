# `Iterative Solution`

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None

def PreOrder(root):
    if root is None:
        return []
    
    stack = [root]
    result = []

    while stack:
        node = stack.pop()

        # 1. Visit root
        result.append(node.data)

        # 2. Push right first
        if node.right:
            stack.append(node.right)
        
        # 3. Push left after pushing right
        if node.left:
            stack.append(node.left)
    
    return result
```

```python
# Driver Code
if __name__ == "__main__":
    #        1
    #       / \
    #      2   3
    #     / \   \
    #    4   5   6

    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)
    root.left.left = Node(4)
    root.left.right = Node(5)
    root.right = Node(3)
    root.right.right = Node(6)

    result = PreOrder(root)
    print(" -> ".join(map(str,result)))
```

# `Recursion`

```python
def preorder(root):
    if root is None:
        return []
    
    result = []

    result.append(root.data)   # Root
    preorder(root.left)        # Left
    preorder(root.right)       # Right

    return result

# Driver Code
if __name__ == "__main__":
    #        1
    #       / \
    #      2   3
    #     / \   \
    #    4   5   6

    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)
    root.left.left = Node(4)
    root.left.right = Node(5)
    root.right = Node(3)
    root.right.right = Node(6)

    result = PreOrder(root)
    print(" -> ".join(map(str,result)))
```
