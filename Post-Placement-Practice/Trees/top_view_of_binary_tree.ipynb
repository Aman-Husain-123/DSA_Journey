```python
from collections import deque


# Node class
class Node:
    
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None


class Solution:
    
    # Function to return Top View of Binary Tree
    def topView(self, root):
        
        # Stores final answer
        ans = []
        
        # Edge case
        if root is None:
            return ans
        
        # Dictionary to store first node
        # for each vertical line
        mp = {}
        
        # Queue for BFS traversal
        # Stores (node, vertical_line)
        q = deque()
        
        # Push root node with line 0
        q.append((root, 0))
        
        # Level order traversal
        while q:
            
            # Pop front element
            node, line = q.popleft()
            
            # Store node if line is seen first time
            if line not in mp:
                mp[line] = node.data
            
            # Push left child
            if node.left:
                q.append((node.left, line - 1))
            
            # Push right child
            if node.right:
                q.append((node.right, line + 1))
        
        # Store nodes in sorted order of lines
        for key in sorted(mp):
            ans.append(mp[key])
        print(mp)
        
        return ans


# ---------------- DRIVER CODE ----------------

# Create binary tree
root = Node(1)

root.left = Node(2)
root.right = Node(3)

root.left.left = Node(4)
root.left.right = Node(5)

root.right.left = Node(6)
root.right.right = Node(7)


# Create object of Solution
obj = Solution()

# Get top view
result = obj.topView(root)

# Print answer
print("Top View of Binary Tree:")

for i in result:
    print(i, end=" ")
```
