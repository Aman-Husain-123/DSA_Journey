```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None


        #               A
        #              / \
        #             B   C
        #            / \
        #           D   E
        #                \
        #                 F    

A = TreeNode('A')
B = TreeNode('B')
C = TreeNode('C')
D = TreeNode('D')
E = TreeNode('E')
F = TreeNode('F')

# Connecting edges with nodes
A.left = B
A.right = C
B.left = D
B.right = E
E.right = F
```
