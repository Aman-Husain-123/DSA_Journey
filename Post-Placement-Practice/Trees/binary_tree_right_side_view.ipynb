```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None

# TODO: Nodes
A = TreeNode("A")
B = TreeNode("B")
C = TreeNode("C")
D = TreeNode("D")
E = TreeNode("E")

# TODO: Edges
A.left = B
A.right = C
B.left = E
B.right = D
```

```python
from collections import deque
def rightsideview(root):
    if not root:
        return []
        
    q = deque([root])
    ans = []

    while q:
        level_size = len(q)

        for i in range(level_size):
            node = q.popleft()

            # Last node of current level
            if i == level_size - 1:
                ans.append(node.val)
            
            if node.left:
                q.append(node.left)
            
            if node.right:
                q.append(node.right)
                
    return ans
    
```

```python
rightsideview(A)
```

# `Using Approach`

```python
from collections import deque

def rightsideview(root):
    if not root:
        return []
    
    queue = deque([root])
    result = []

    while queue:
        size = len(queue)
        last_seen = None

        for _ in range(size):
            node = queue.popleft()
            last_seen = node.val

            if node.left:
                queue.append(node.left)
            
            if node.right:
                queue.append(node.right)
        
        result.append(last_seen)
    return result
```

```python
rightsideview(A)
```

# `Recursive Approach`

```python
# TODO: SC :-O(n)
#       TC :- O(H)  Space is generally consider in an algorithm when returning answer 


# Reversal of pre-order travsersal is used
# Root left right   ->  Root right left
def recursive_approach(root):
    ans = []

    def dfs(node,level):
        if node is not None:
            return
        
        if level == len(ans):
            ans.append(node.val)    
        
        dfs(node.right,level + 1)
        dfs(node.left,level + 1)

    
    dfs(root,0)

    return ans
```

```python
recursive_approach(A)
```
