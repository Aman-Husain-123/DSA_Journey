```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None

# TODO: Nodes
A = TreeNode(1)
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)

# TODO: Edges
A.left = B
A.right = C
B.left = E
B.right = D
```

```python
from collections import deque

def rightsideview(root):
    if not root:
        return []
    
    queue = deque([root])
    result = []

    while queue:
        size = len(queue)
        last_seen = None

        for _ in range(size):
            node = queue.popleft()
            last_seen = node.val

            if node.left:
                queue.append(node.left)
            
            if node.right:
                queue.append(node.right)
        
        result.append(last_seen)
    return result
```
