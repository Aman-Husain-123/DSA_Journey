```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
```

```python
# TODO: Nodes
A = TreeNode("A")
B = TreeNode("B")
C = TreeNode("C")
D = TreeNode("D")
E = TreeNode("E")
F = TreeNode("F")
```

```python
# TODO: Edges
A.left = B
A.right = C
B.left = D
B.right = E
C.left = F
```

```python
def post_order_traversal(node):
    if not node:
        return
    
    post_order_traversal(node.left)
    post_order_traversal(node.right)
    print(node.val,end = " ")
```

```python
post_order_traversal(A)
```
