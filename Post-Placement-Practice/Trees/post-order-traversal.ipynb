```python
def post_order_traversal(node):
    if not node:
        return
    
    post_order_traversal(node.left)
    post_order_traversal(node.right)
    print(node.val,end = " ")
```

# `Iterative Apporach`

```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
```

```python
# TODO: Nodes
A = TreeNode("A")
B = TreeNode("B")
C = TreeNode("C")
D = TreeNode("D")
E = TreeNode("E")
F = TreeNode("F")
```

```python
# TODO: Edges
A.left = B
A.right = C
B.left = D
B.right = E
C.left = F
```

```python
def post_order_approach(root):
    stack1 = []
    stack2 = []
    stack1.append(root)

    if not root:
        return[]
    
    while (len(stack1) > 0):
        curr = stack1.pop()
        stack2.append(curr)

        if curr.left is not None:
            stack1.append(curr.left)
        
        if curr.right is not None:
            stack1.append(curr.right)
    
    while (len(stack2) > 0):
        data = stack2.pop()
        print(data.val)
```

```python
post_order_approach(A)  
```
