```python
# define tree node structure
class TreeNode:
    # constructor
    def __init__(self, x: int):
        # value field
        self.val = x
        # left child reference
        self.left = None
        # right child reference
        self.right = None
```

```python
# solution class
class Solution:
    # keep same field as original (unused in iterative version)
    def __init__(self):
        self.prev = None
    
    # flatten binary tree to right-skewed linked list (iterative)
    def flatten(self,root:'TreeNode') -> None:
        # handle empty tree
        if root is None:
            return
        # create stack for DFS
        st = []
        # push root to start traversal
        st.append(root)
        # process while stack has nodes
        while st:
            # take top node
            cur = st.pop()
            # push right child first (so left is processed first)
            if cur.right is not None:
                st.append(cur.right)
            # push left child next
            if cur.left is not None:
                st.append(cur.left)
            # if stack not empty, link current's right to next node
            if st:
                cur.right = st[-1]
            # nullify left to form right-only chain
            cur.left = None
```

```python
def print_preorder(root:'TreeNode') -> None:
    # stop at null
    if root is None:
        return
    
    print(root.val,end = " ")
    print_preorder(root.left)
    print_preorder(root.right)

# print along right pointers
def print_flatten_tree(root: 'TreeNode') -> None:
    # stop at null
    if root is None:
        return
    
    print(root.val,end = " ")
    print_flatten_tree(root.right)
```

```python
# program entry
def main():
    # build sample tree
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)
    root.left.left = TreeNode(4)
    root.left.right = TreeNode(5)
    root.left.right.right = TreeNode(6)
    root.right.right = TreeNode(7)
    root.right.left = TreeNode(8)

    # create solver
    sol = Solution()

    # show preorder before flatten
    print("Binary Tree Preorder: ", end="")
    print_preorder(root)
    print()

    # perform flatten
    sol.flatten(root)

    # show right-chain after flatten
    print("Binary Tree After Flatten: ", end="")
    print_flatten_tree(root)
    print()

# run main
if __name__ == "__main__":
    main()
```
