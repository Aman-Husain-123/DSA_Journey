```python
# define tree node structure
class TreeNode:
    # constructor to initialize node
    def __init__(self, x: int):
        # store value
        self.val = x
        # left child reference
        self.left = None
        # right child reference
        self.right = None
```

```python
# Solution Class Holding State
class Solution:
    # initialize previous node reference
    def __init__(self):
        self.prev = None

    # flatten tree to right-skewed linked list (preorder sequence)
    def flatten(self, root: TreeNode) -> None:
        # Return if node is null
        if root is None:
            return
        
        # flatten right subtree first
        self.flatten(root.right)

        # flatten left subtree first
        self.flatten(root.left)

        # connect current node's right to previously processed node
        root.right = self.prev

        # nullify left pointer
        root.left = None

        # update previous to current
        self.prev = root
```

```python
# print preorder traversal
def print_preorder(root: TreeNode) -> None:
    # return on null
    if root is None:
        return
    # print current node
    print(root.val, end=" ")
    # traverse left subtree
    print_preorder(root.left)
    # traverse right subtree
    print_preorder(root.right)
```

```python
# print nodes following right pointers
def print_flatten_tree(root: TreeNode) -> None:
    # return on null
    if root is None:
        return
    # print current node
    print(root.val, end=" ")
    # move to right child
    print_flatten_tree(root.right)
```

```python
# program entry point
def main():
    # construct sample tree
    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(3)
    root.left.left = TreeNode(4)
    root.left.right = TreeNode(5)
    root.left.right.right = TreeNode(6)
    root.right.right = TreeNode(7)
    root.right.left = TreeNode(8)

    # create solution instance
    sol = Solution()

    # print original tree preorder
    print("Binary Tree Preorder: ", end="")
    print_preorder(root)
    print()

    # flatten the tree
    sol.flatten(root)

    # print flattened tree via right pointers
    print("Binary Tree After Flatten: ", end="")
    print_flatten_tree(root)
    print()

# run main
if __name__ == "__main__":
    main()
```
