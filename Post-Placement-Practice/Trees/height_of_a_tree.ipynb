```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None


# TODO: Nodes
A = TreeNode("A")
B = TreeNode("B")
C = TreeNode("C")
D = TreeNode("D")
E = TreeNode("E")

A.left = B
A.right = C
B.left = D
D.left = E
```

```python
                        #         A
                        #        / \
                        #       B   C
                        #      /
                        #     D
                        #    /
                        #   E  
```

```python
def height(root):
    # TODO: Base Case
    if root is None:
        return 0
    
    # TODO: Find height of the left subtree
    leftheight = height(root.left)
    

    # TODO: Find height of the right subtree
    rightheight = height(root.right)
    

    return max(leftheight,rightheight) + 1    
```
