```python
from collections import deque
```

```python
def level_order_traversal(node):
    if not node:
        return
    
    q = deque([node])
    while q:
        curr = q.popleft()
        print(curr.val,end = ' ')

        if curr.left:
            q.append(node.left)
        
        if curr.right:
            q.append(node.right)
```

```python
level_order_traversal(A)
```
