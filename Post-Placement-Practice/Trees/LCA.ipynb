```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None


        #               A
        #              / \
        #             B   C
        #            / \
        #           D   E
        #                \
        #                 F    

A = TreeNode('A')
B = TreeNode('B')
C = TreeNode('C')
D = TreeNode('D')
E = TreeNode('E')
F = TreeNode('F')

# Connecting edges with nodes
A.left = B
A.right = C
B.left = D
B.right = E
E.right = F
```

```python
def LCA(root,p,q):
    ans = None
    def dfs(root):
        nonlocal ans

        if root is None:
            return 0
        
        left = dfs(root.left)
        right = dfs(root.right)


        selff = 0
        if root == p or root == q:
            selff = 1

        total = selff + left + right
        if total >= 2 and ans is None:
                ans = root
            
        return 1 if total > 0 else 0
    dfs(root)
    return ans
```

```python
result = LCA(A,D,F)
print(f'LCA of D and F is: {result.val}')
```

# Pratyush Solution

```python
ans = None
def func(node,p,q):
    global ans
    if node is None:
        return 0
    
    left = func(node.left,p,q)
    right = func(node.right,p,q)

    selff = 0

    if node == p or node == q:
        selff = 1
    
    total = selff + left + right
    if total >= 2 and ans is None:
        ans = node
    
    return 1 if total > 0 else 0
```

```python
func(A,D,F)
```

# `PW Solution`

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, x):
#         self.val = x
#         self.left = None
#         self.right = None

class Solution:
    def exists(self,root,target):
        if root == None:
            return False
        
        if root == target:
            return True
        
        return self.exists(root.left,target) or self.exists(root.right,target)
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        if root == p or root == q:
            return root
        
        elif self.exists(root.left,p) and self.exists(root.right,q):
            return root
        elif self.exists(root.right,p) and self.exists(root.left,q):
            return root
        elif self.exists(root.left,p) and self.exists(root.left,q):
            return self.lowestCommonAncestor(root.left,p,q)
        else:
            return self.lowestCommonAncestor(root.right,p,q)
        
    
```
