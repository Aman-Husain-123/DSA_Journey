```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None


        #               A
        #              / \
        #             B   C
        #            / \
        #           D   E
        #                \
        #                 F    

A = TreeNode('A')
B = TreeNode('B')
C = TreeNode('C')
D = TreeNode('D')
E = TreeNode('E')
F = TreeNode('F')

# Connecting edges with nodes
A.left = B
A.right = C
B.left = D
B.right = E
E.right = F
```

```python
def LCA(root,p,q):
    ans = None
    def dfs(root):
        nonlocal ans

        if root is None:
            return 0
        
        left = dfs(root.left)
        right = dfs(root.right)


        selff = 0
        if root == p or root == q:
            selff = 1

        total = selff + left + right
        if total >= 2 and ans is None:
                ans = root
            
        return 1 if total > 0 else 0
    dfs(root)
    return ans
```

```python
result = LCA(A,D,F)
print(f'LCA of D and F is: {result.val}')
```
