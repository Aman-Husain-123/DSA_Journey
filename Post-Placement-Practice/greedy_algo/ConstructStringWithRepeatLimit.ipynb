```python
import heapq
from collections import Counter


class Solution:
    def repeatLimitedString(self, s: str, repeatLimit: int) -> str:
        # Count frequency of each character
        freq = Counter(s)

        # Max Heap (Python has min heap, so use negative ASCII values)
        pq = []
        for ch, count in freq.items():
            heapq.heappush(pq, (-ord(ch), ch, count))

        result = []

        while pq:
            _, ch1, cnt1 = heapq.heappop(pq)

            length = min(repeatLimit, cnt1)

            # Add current largest character
            result.append(ch1 * length)

            cnt1 -= length

            # If some occurrences are still left
            if cnt1 > 0:
                if not pq:
                    break

                _, ch2, cnt2 = heapq.heappop(pq)

                # Insert one occurrence of second largest character
                result.append(ch2)

                cnt2 -= 1

                # Push back second largest if still available
                if cnt2 > 0:
                    heapq.heappush(pq, (-ord(ch2), ch2, cnt2))

                # Push back current largest
                heapq.heappush(pq, (-ord(ch1), ch1, cnt1))

        return "".join(result)R


# ---------------- Driver Code ----------------

s = "cczazcc"
repeatLimit = 3

obj = Solution()
print(obj.repeatLimitedString(s, repeatLimit))
```
