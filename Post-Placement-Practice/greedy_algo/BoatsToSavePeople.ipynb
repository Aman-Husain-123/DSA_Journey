```python
def BoatsToSavePeople(people,limit):
    people.sort()

    i = 0
    j = len(people) - 1

    bt = 0
    while i <= j:
        if people[i] + people[j] <= limit:
            bt += 1
            i += 1
            j -= 1
        else:
            bt += 1
            j -= 1
    
    return bt

people = [1,2] 
limit = 3
BoatsToSavePeople(people,limit)
```
