```python
class Solution:
    def maximumUnits(self, boxTypes, truckSize):
        # Sort by units per box in descending order
        boxTypes.sort(key=lambda x: x[1], reverse=True)
        print(boxTypes)
        profit = 0

        for boxes, units in boxTypes:
            if boxes <= truckSize:
                profit += boxes * units
                truckSize -= boxes
            else:
                profit += truckSize * units
                truckSize = 0

            if truckSize == 0:
                break

        return profit


# ---------------- Driver Code ----------------

boxTypes = [[1, 3], [2, 2], [3, 1]]
truckSize = 4

obj = Solution()
ans = obj.maximumUnits(boxTypes, truckSize)

print("Maximum Units:", ans)
```
