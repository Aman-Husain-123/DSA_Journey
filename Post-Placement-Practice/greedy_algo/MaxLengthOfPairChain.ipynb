```python
from typing import List
class Solution:
    def findLongestChain(self, pairs: List[List[int]]) -> int:
        pairs.sort(key = lambda x : x[1])
        ans = 0
        last_interval = pairs[0][1]

        for i in range(1,len(pairs)):
            if pairs[i][0] > last_interval:
                last_interval = pairs[i][1]
                ans+=1
            # else:
            #     continue
        
        return ans+1
```

```python
pairs = [[1,2],[2,3],[3,4]]
obj = Solution()
obj.findLongestChain(pairs)
```
