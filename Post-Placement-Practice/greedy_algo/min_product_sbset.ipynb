```python
def minProductSubset(arr):
    cn = 0
    cz = 0
    cp = 0
    prod_pos = 1
    prod_neg = 1
    largest_neg = float('-inf')

    for i in range(len(arr)):
        if arr[i] < 0:
            cn += 1
            prod_neg *= arr[i]
            largest_neg = max(largest_neg,arr[i])
        if arr[i] == 0:
            cz += 1
        if arr[i] > 0:
            cp += 1
            prod_pos *= arr[i]
    
    if cn == 0:
        if cz > 0:
            return 0
        else:
            it = min(arr)
            return it
    else:
        if cn % 2 == 0:
            return (prod_neg / largest_neg) * prod_pos
        else:
            return prod_neg * prod_pos

arr = [-2,-3,1,4,-2]
minProductSubset(arr)
```
