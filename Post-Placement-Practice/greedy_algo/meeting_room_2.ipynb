```python
class Solution:
    def minMeetingRooms(self, intervals):
        start = []
        ending = []

        # Store all start and end times separately
        for s, e in intervals:
            start.append(s)
            ending.append(e)

        # Sort both arrays
        start.sort()
        ending.sort()

        ans = 0
        rooms = 0
        i = 0
        j = 0

        while i < len(start) and j < len(ending):
            if start[i] < ending[j]:
                # A meeting is starting
                rooms += 1
                ans = max(ans, rooms)
                i += 1

            elif start[i] > ending[j]:
                # A meeting has ended
                rooms -= 1
                j += 1

            else:
                # One meeting ends exactly when another starts
                i += 1
                j += 1

        return ans


# ---------------- Driver Code ----------------

intervals = [
    [0, 30],
    [5, 10],
    [15, 20]
]

obj = Solution()
print("Minimum Meeting Rooms:", obj.minMeetingRooms(intervals))
```
