```python
class Solution:
    def brokenCalc(self, startValue: int, target: int) -> int:
        ans = 0
        while target > startValue:
            if target % 2 == 0:
                target //= 2
            else:
                target += 1
            
            ans += 1
        
        ans += startValue - target
        return ans
```

```python
startValue = 2
target = 3
obj = Solution()
obj.brokenCalc(startValue, target)
```
