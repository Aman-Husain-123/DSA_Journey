```python
from typing import List
class Solution:
    def maximumUnits(self, boxTypes: List[List[int]], truckSize: int) -> int:
        # Sort by units per box in descending order
        # TC :- N(log n)
        # SC :- O(n)
        boxTypes.sort(key=lambda x: x[1], reverse=True)
        profit = 0

        for boxes, units in boxTypes:
            if boxes <= truckSize:
                profit += boxes * units
                truckSize -= boxes
            else:
                profit += truckSize * units
                truckSize = 0

            if truckSize == 0:
                break

        return profit
```

```python
boxTypes = [[1,3],[2,2],[3,1]]
truckSize = 4
obj = Solution()
obj.maximumUnits(boxTypes, truckSize)
```
