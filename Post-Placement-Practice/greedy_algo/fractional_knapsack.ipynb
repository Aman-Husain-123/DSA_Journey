```python
class Solution:
    def fractionalKnapsack(self, profit, weights, n, W):
        arr = []

        # Store (profit, weight)
        for i in range(n):
            arr.append((profit[i], weights[i]))

        # Sort by profit/weight ratio in descending order
        arr.sort(key=lambda x: x[0] / x[1], reverse=True)

        result = 0.0

        for p, wt in arr:
            if wt <= W:
                result += p
                W -= wt
            else:
                result += (p / wt) * W
                W = 0
                break

        return result


# ---------------- Driver Code ----------------

profit = [60, 100, 120]
weights = [10, 20, 30]
W = 50
n = len(profit)

obj = Solution()
ans = obj.fractionalKnapsack(profit, weights, n, W)

print("Maximum Profit:", ans)
```
