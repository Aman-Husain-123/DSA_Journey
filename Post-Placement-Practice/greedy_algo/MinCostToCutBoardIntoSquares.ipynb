```python
def minCostToBreakGrid(n, m, vertical, horizontal):
    vertical.sort(reverse=True)
    horizontal.sort(reverse=True)

    hz = 1
    vr = 1

    h = 0
    v = 0

    ans = 0

    while h < len(horizontal) and v < len(vertical):
        if vertical[v] > horizontal[h]:
            ans += vertical[v] * vr
            hz += 1
            v += 1
        else:
            ans += horizontal[h] * hz
            vr += 1
            h += 1

    while h < len(horizontal):
        ans += horizontal[h] * hz
        vr += 1
        h += 1

    while v < len(vertical):
        ans += vertical[v] * vr
        hz += 1
        v += 1

    return ans
```

```python
n = 4
m = 6
x = [2, 1, 3, 1, 4]
y = [4, 1, 2]

minCostToBreakGrid(n,m,x,y)
```
