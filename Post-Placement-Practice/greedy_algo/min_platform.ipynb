```python
class Solution:    
    def minPlatform(self, arr, dep):
        # code here
        arr.sort()
        dep.sort()
        
        i = 1
        j = 0
        
        platform = 1
        ans = 1
        
        while i < len(arr) and j < len(dep):
            if arr[i] <= dep[j]:
                platform += 1
                ans = max(ans,platform)
                i += 1
            else:
                platform -= 1
                j += 1
        
        return ans
        
```

```python
arr= [900, 940, 950, 1100, 1500, 1800]
dep = [910, 1200, 1120, 1130, 1900, 2000]

obj = Solution()
obj.minPlatform(arr, dep)
```
