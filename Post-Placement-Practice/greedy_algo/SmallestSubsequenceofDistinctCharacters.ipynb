```python
from collections import Counter
class Solution:
    def smallestSubsequence(self, s: str) -> str:
        # Approach
        # Traverse string
        # Keep 3 DS (freq,set,stack)
            # freq :- Tells if we have an extra freq through which i can accomodate this char later on
            # visited(set) :- If a char has been taken or not
            # stack :- keep a track of last char

        freq = Counter(s)
        vis = set()
        stack = []

        for ch in s:
            freq[ch] -= 1

            if ch in vis:
                continue

            while (
                stack
                and stack[-1] > ch
                and freq[stack[-1]] > 0
            ):
                vis.remove(stack.pop())

            stack.append(ch)
            vis.add(ch)

        return "".join(stack)



# This problem has four key ideas:

# 1. Each distinct character appears exactly once.
# 2. The answer must be a subsequence, so you cannot rearrange characters.
# 3. Among all valid subsequences, choose the lexicographically smallest.
# 4. While scanning left to right, you may remove previously chosen characters only if:
#    - the current character is smaller, and
#    - the removed character appears again later.
```

```python
s = "cbacdcbc"
obj = Solution()
obj.smallestSubsequence(s)
```
