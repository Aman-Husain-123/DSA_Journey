```python
import heapq

class Solution:
    def largestSumAfterKNegations(self, nums, k):
        min_heap = []
        summ = 0

        # Build min heap and calculate initial sum
        for num in nums:
            heapq.heappush(min_heap, num)
            summ += num

        # Perform k negations
        while k > 0:
            ele = heapq.heappop(min_heap)

            # If smallest element is 0, further negations won't change the sum
            if ele == 0:
                break

            k -= 1

            # Update the sum after negation
            summ -= ele
            ele = -ele
            summ += ele

            # Push the negated element back into the heap
            heapq.heappush(min_heap, ele)

        return summ


# ---------------- Driver Code ----------------

nums = [4, 2, 3]
k = 1

obj = Solution()
ans = obj.largestSumAfterKNegations(nums, k)

print("Array:", nums)
print("K:", k)
print("Maximum Sum:", ans)
```
