```python
class Solution:
    def numRabbits(self, answers):
        # TC and SC -- > O(n)
        mp = {}
        result = 0

        for ans in answers:
            group_size = ans + 1

            if group_size not in mp:
                # Start a new color group
                result += group_size

                # If answer is 0, group is already complete
                if ans == 0:
                    continue

                mp[group_size] = 1

            else:
                mp[group_size] += 1

                # Group is complete
                if mp[group_size] == group_size:
                    del mp[group_size]

        return result
```
