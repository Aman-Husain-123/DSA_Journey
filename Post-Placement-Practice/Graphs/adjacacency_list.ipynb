```python
# Number of vertices
n = 5

# Edge list (undirected graph)
edges = [
    [0, 1],
    [0, 2],
    [1, 3],
    [1, 4],
    [2, 4]
]

# Initialize adjacency list
adj = [[] for _ in range(n)]

# Add edges
for src, dest in edges:
    print(src,end=" ")
    print(dest)
    adj[src].append(dest)
    adj[dest].append(src)  # Remove this line for directed graph

# Print adjacency list
for i in range(n):
    print(f"{i}: {adj[i]}")
```

```python
mat = {}
edges = [[0, 1], [0, 2], [1, 3], [2, 4]]


for edge in edges:
    src = edge[0]
    dest = edge[1]

    if src not in mat:
        mat[src] = {}
    if dest not in mat:
        mat[dest] = {}

    mat[src][dest] = 1
    mat[dest][src] = 1
mat
```
