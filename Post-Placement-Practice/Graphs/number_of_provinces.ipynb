## Number of Provinces — Graph Visualization

**Problem:** Given an `n x n` adjacency matrix `isConnected`, find the total number of **provinces** (connected components).

**Example:** `isConnected = [[1,1,0],[1,1,0],[0,0,1]]`

Graph

- **Province 1:** Nodes 0 and 1 are directly connected → form one province.
- **Province 2:** Node 2 has no connections → forms its own province.
- **Answer:** `2` provinces

**Approach:** Run BFS from every unvisited node. Each BFS traversal visits an entire province. Count the number of times BFS is triggered.

```python
from collections import deque
def bfs(i, vis, adj):
    n = len(adj)

    vis[i] = True
    q = deque([i])

    while q:
        front = q.popleft()

        for j in range(n):
            if adj[front][j] == 1 and not vis[j]:
                vis[j] = True
                q.append(j)
```

```python
def findCircleNum(isConnected):
    n = len(isConnected)
    count = 0
    visited = [False] * n
    
    for i in range(n):
        if not visited[i]:
            bfs(i, visited, isConnected)  # marks all nodes in this province as visited
            count += 1                    # each BFS call = one new province

    return count


isConnected = [[1,1,0],[1,1,0],[0,0,1]]
print(findCircleNum(isConnected))
```
