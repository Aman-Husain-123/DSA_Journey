# `Breadth First Traversal`

```python
# Method Definition
from collections import deque
queue = deque()

def BFT(graph,visited,node):
    visited.add(node)

    # Queue - FIFO
    queue.appendleft(node)

    while queue:
        node = queue.pop()
        print(node,end = " ")

        for adjacent_node in graph[node]:
            if adjacent_node not in visited:
                visited.add(adjacent_node)
                queue.appendleft(adjacent_node)
```

```python
# Driver Code
graph = {
    'A' : ['B','C','D'],
    'B' : ['E'],
    'C' : ['E','F'],
    'D' : ['F'],
    'E' : ['G'],
    'F' : ['G'],
    'G' : []
}

# Visited set to take care of all the visited nodes
visited = set()

# Function Calling
BFT(graph,visited,'A')
```

# `Another Approach`

```python
def BFTT(start_node,graph):
    queue = deque([start_node])
    visited = set()
    res = []

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            res.append(node)

            neighbours = graph[node]
            for nei in neighbours:
                if nei not in visited:
                    queue.append(nei)
    
    return res
```

```python
# Driver Code
graph = {
    'A' : ['B','C','D'],
    'B' : ['E'],
    'C' : ['E','F'],
    'D' : ['F'],
    'E' : ['G'],
    'F' : ['G'],
    'G' : []
}

# Function Calling
print(BFTT('A', graph))
```

# `Using level order`

```python
def BFTT(start_node,graph):
    queue = deque([start_node])
    visited = set([start_node])
    res = []

    while queue:
        size = len(queue)
        level = []
        for _ in range(size):
            node = queue.popleft()
            level.append(node)
            neighbours = graph[node]
            for nei in neighbours:
                if nei not in visited:
                    visited.add(nei)
                    queue.append(nei)
        res.append(level)
    
    return res
```

```python
# Driver Code
graph = {
    'A' : ['B','C','D'],
    'B' : ['E'],
    'C' : ['E','F'],
    'D' : ['F'],
    'E' : ['G'],
    'F' : ['G'],
    'G' : []
}

# Function Calling
BFTT('A', graph)
```

# `Iterative Approach`

```python
from collections import deque
def BFTT(start_node,graph):
    queue = deque([start_node])
    visited = set()
    res = []

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            res.append(node)

            neighbours = graph[node]
            for nei in neighbours:
                if nei not in visited:
                    queue.append(nei)
    return res
```

```python
# Driver Code
graph = {
    'A' : ['B','C','D'],
    'B' : ['E'],
    'C' : ['E','F'],
    'D' : ['F'],
    'E' : ['G'],
    'F' : ['G'],
    'G' : []
}

# Function Calling
BFTT('A', graph)
```
