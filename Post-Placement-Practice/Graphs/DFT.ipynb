# `Depth First Traversal`

```python
# Method Definition For Depth First Traversal
def depth_first_traversal(visited,graph,node):
    if node not in visited:
        print(node,end = " ")
        visited.add(node)
    
    # Checking for adjacent nodes
    for adjacent_node in graph[node]:
        # Recursive call
        depth_first_traversal(visited,graph,adjacent_node)


# Visited set to keep track of all visited nodes
visited = set()

## driver code
graph = {
    '5' : ['3','7'],
    '3' : ['2', '4'],
    '7' : ['8'],
    '2' : [],
    '4' : ['8'],
    '8' : []
}

depth_first_traversal(visited, graph, '5')
```

# `Iterative Approach`

```python
def dfs(start_node,graph):
    visited = set()
    res = []
    stack = [start_node]

    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            res.append(node)
            neighbours = graph[node]

            for nei in neighbours:
                if nei not in visited:
                    stack.append(nei)
    return res
```

```python
## driver code
graph = {
    '5' : ['3','7'],
    '3' : ['2', '4'],
    '7' : ['8'],
    '2' : [],
    '4' : ['8'],
    '8' : []
}

print(dfs('5', graph))
```
