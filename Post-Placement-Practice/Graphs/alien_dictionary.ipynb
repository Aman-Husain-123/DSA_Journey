```python
from collections import deque

class Solution:
    def topoSort(self, V, adj):
        # Step 1: Calculate indegree
        indegree = [0] * V

        for i in range(V):
            for node in adj[i]:
                indegree[node] += 1

        # Step 2: Push all vertices with indegree 0
        q = deque()

        for i in range(V):
            if indegree[i] == 0:
                q.append(i)

        # Step 3: Kahn's Algorithm
        topo = []

        while q:
            node = q.popleft()
            topo.append(node)

            for neighbor in adj[node]:
                indegree[neighbor] -= 1

                if indegree[neighbor] == 0:
                    q.append(neighbor)

        return topo

    def findOrder(self, dict, N, K):
        # Create graph
        adj = [[] for _ in range(K)]

        # Compare adjacent words
        for i in range(N - 1):
            s1 = dict[i]
            s2 = dict[i + 1]

            length = min(len(s1), len(s2))

            for j in range(length):
                if s1[j] != s2[j]:
                    adj[ord(s1[j]) - ord('a')].append(ord(s2[j]) - ord('a'))
                    break

        # Perform Topological Sort
        topo = self.topoSort(K, adj)

        # Convert indices back to characters
        ans = ""

        for node in topo:
            ans += chr(node + ord('a'))

        return ans


# ---------------- Driver Code ----------------

dictionary = [
    "baa",
    "abcd",
    "abca",
    "cab",
    "cad"
]

N = len(dictionary)
K = 4

obj = Solution()
print(obj.findOrder(dictionary, N, K))
```
