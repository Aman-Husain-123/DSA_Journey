```python
class Solution:
    
    def dfs(self, node, adj, visited, path):
        visited[node] = True
        path[node] = True

        for neighbor in adj[node]:

            # Cycle found
        A    if path[neighbor]:
                return True

            # Visit unvisited neighbor
            if not visited[neighbor]:
                if self.dfs(neighbor, adj, visited, path):
                    return True

        # Backtrack
        path[node] = False
        return False

    def isCyclic(self, V, adj):
        visited = [False] * V
        path = [False] * V

        for i in range(V):
            if not visited[i]:
                if self.dfs(i, adj, visited, path):
                    return True

        return False
```
