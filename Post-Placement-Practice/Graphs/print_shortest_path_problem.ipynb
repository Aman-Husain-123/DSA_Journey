```python
import heapq

class Solution:

    def shortestPath(self, n, m, edges):

        # Step 1: Build Adjacency List
        adj = [[] for _ in range(n + 1)]

        for u, v, w in edges:
            adj[u].append((v, w))
            adj[v].append((u, w))

        # Step 2: Initialize Distance and Parent arrays
        dist = [float('inf')] * (n + 1)
        parent = [i for i in range(n + 1)]

        dist[1] = 0

        # Step 3: Min Heap
        pq = []
        heapq.heappush(pq, (0, 1))

        # Step 4: Dijkstra
        while pq:

            dis, node = heapq.heappop(pq)

            if dis > dist[node]:
                continue

            for adjNode, edgeWeight in adj[node]:

                if dis + edgeWeight < dist[adjNode]:

                    dist[adjNode] = dis + edgeWeight
                    parent[adjNode] = node

                    heapq.heappush(pq, (dist[adjNode], adjNode))

        # Step 5: If destination is unreachable
        if dist[n] == float('inf'):
            return [-1]

        # Step 6: Reconstruct Path
        path = []

        node = n

        while parent[node] != node:
            path.append(node)
            node = parent[node]

        path.append(1)

        path.reverse()

        return path


# ---------------- Driver Code ----------------

n = 5
m = 6

edges = [
    [1, 2, 2],
    [2, 5, 5],
    [1, 3, 4],
    [3, 4, 3],
    [4, 5, 1],
    [2, 3, 1]
]

obj = Solution()

path = obj.shortestPath(n, m, edges)

print("Shortest Path =", path)
```
