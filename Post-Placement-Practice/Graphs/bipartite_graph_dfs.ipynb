```python
from collections import deque
class Solution:
    def dfs(self,node,col,color,adj):
        color[node] = col
        for i in adj[node]:
            if color[i] == -1:
                if self.dfs(i,col,color,adj) == False:
                    return False
                elif color[i] == col:
                    return False
        return True
    def isBipartite(self,V,adj):
        color = [-1] * V
        for i in range(V):
            if color[i] == -1:
                if self.dfs(i,0,color,adj) == False:
                    return False
        
        return True
```

```python
V = 4

adj = [
    [1, 3],   # 0
    [0, 2],   # 1
    [1, 3],   # 2
    [0, 2]    # 3
]

obj = Solution()
obj.isBipartite(V,adj)
```
