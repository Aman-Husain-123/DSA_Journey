```python
from collections import defaultdict

class Graph:
    def __init__(self):
        # adjacency list
        # {node: [(neighbor, weight), ...]}
        self.adj = defaultdict(list)

    def addEdge(self, u, v, weight):
        self.adj[u].append((v, weight))

    def printAdj(self):
        for node in self.adj:
            print(f"{node} -> ", end="")
            for neighbor, weight in self.adj[node]:
                print(f"({neighbor}, {weight})", end=", ")
            print()



# Driver Code
g = Graph()

g.addEdge(0, 1, 5)
g.addEdge(0, 2, 3)
g.addEdge(1, 2, 2)
g.addEdge(1, 3, 6)
g.addEdge(2, 3, 7)
g.addEdge(2, 4, 4)
g.addEdge(2, 5, 2)
g.addEdge(3, 4, -1)
g.addEdge(4, 5, -2)

g.printAdj()
n = 6
vis = [] * n
for i in range(n):
    if not vis[i]
```
