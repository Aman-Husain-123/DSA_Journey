```python
from collections import deque

class Solution:
    # Function to check cycle starting from a source node using BFS
    def checkForCycle(self,src,V,adj,vis):
        # Mark the source node as visited
        vis[src] = True

        # Queue will store (node,parent) pairs
        q = deque()
        q.append((src,-1)) # Source node has no parent

        # Perform BFS
        while q:
            node,parent = q.popleft()

            # Traverse all adjancent nodes
            for adjacentNode in adj[node]:
                # If adjacent node is not visited, mark it visited
                # and push it into queue with current node as parent
                if not vis[adjacentNode]:
                    vis[adjacentNode] = True
                    q.append((adjacentNode,node))
                
                # If adjacent node is already visited and is not the parent,
                # then a cycle exists
                elif parent != adjacentNode:
                    return True
        # No cycle found
        return False
    # Function to detect cycle in an undirected graph
    def isCycle(self, V, adj):
        # Visited array initialized to False
        vis = [False] * V

        # Check every component of the graph
        for i in range(V):

            # Start BFS only if node is unvisited
            if not vis[i]:
                if self.checkForCycle(i, V, adj, vis):
                    return True

        # No cycle in any component
        return False
```

```python
V = 5
adj = [
    [1],        # 0
    [0, 2, 4],  # 1
    [1, 3],     # 2
    [2, 4],     # 3
    [1, 3]      # 4
]

obj = Solution()
print(obj.isCycle(V, adj))   # True
```
