```python
from collections import deque
from typing import List

class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        rows, cols = len(grid), len(grid[0])

        queue = deque()

        total_oranges = 0
        rotten_count = 0
        total_time = 0

        # Traverse the grid and count oranges and store rotten oranges
        for i in range(rows):
            for j in range(cols):
                if grid[i][j] != 0:
                    total_oranges += 1

                if grid[i][j] == 2:
                    queue.append((i, j))

        directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        while queue:
            size = len(queue)
            rotten_count += size
            print(rotten_count)

            made_rotten = False

            for _ in range(size):
                i, j = queue.popleft()

                for dr, dc in directions:
                    ni = i + dr
                    nj = j + dc

                    if (0 <= ni < rows and
                        0 <= nj < cols and
                        grid[ni][nj] == 1):

                        grid[ni][nj] = 2
                        queue.append((ni, nj))
                        made_rotten = True

            if made_rotten:
                total_time += 1

        if rotten_count == total_oranges:
            return total_time

        return -1
```

```python
grid = [
    [2,1,1],
    [1,1,0],
    [0,1,1]
]
obj = Solution()
obj.orangesRotting(grid)
```
