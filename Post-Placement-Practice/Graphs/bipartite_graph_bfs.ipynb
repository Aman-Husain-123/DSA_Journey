```python
from collections import deque

class Solution:

    # BFS function to check whether a connected component
    # starting from 'start' is bipartite or not.
    def check(self, start, V, adj, color):

        # Queue for BFS
        q = deque()

        # Push the starting node into the queue
        q.append(start)

        # Assign the first color (0) to the starting node
        color[start] = 0

        # Perform BFS
        while q:

            # Remove the front node
            node = q.popleft()
            # print(node)
            # Traverse all adjacent nodes
            for neigh in adj[node]:
                # print(neigh)

                # If the adjacent node is not yet colored
                if color[neigh] == -1:

                    # Assign the opposite color
                    color[neigh] = 1 - color[node]

                    # Push it into the queue for further traversal
                    q.append(neigh)

                # If the adjacent node has the same color,
                # then the graph is not bipartite
                elif color[neigh] == color[node]:
                    return False

        # No conflict found
        return True

    # Main function
    def isBipartite(self, V, adj):

        # Initialize all vertices as uncolored (-1)
        color = [-1] * V

        # Check every connected component
        for i in range(V):

            # If the vertex is not yet colored,
            # start a BFS from it
            if color[i] == -1:

                # If any component is not bipartite,
                # return False immediately
                if not self.check(i, V, adj, color):
                    return False

        # All connected components are bipartite
        return True
```

```python
V = 4

adj = [
    [1, 3],   # 0
    [0, 2],   # 1
    [1, 3],   # 2
    [0, 2]    # 3
]

obj = Solution()
obj.isBipartite(V,adj)
```
