# `Kanhs Algorithm(Using BFS)`

```python
from collections import deque
def toposort(V,edges):
    adj = [[] for _ in range(V)]
    indegree = [0] * V

    for u,v in edges:
        adj[u].append(v)
        indegree[v] += 1
    
    q = deque()
    for i in range(V):
        if indegree[i] == 0:
            q.append(i)
    
    ans = []

    # BFS
    while q:
        node = q.popleft()
        ans.append(node)

        for nei in adj[node]:
            indegree[nei] -= 1
            if indegree[nei] == 0:
                q.append(nei)
    
    return ans
```

```python
V = 4
edges = [[3, 0], [1, 0], [2, 0]]
toposort(V,edges)
```
