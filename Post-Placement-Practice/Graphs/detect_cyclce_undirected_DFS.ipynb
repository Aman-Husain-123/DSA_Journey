```python
def dfs(adj, node, parent, vis):
        vis[node] = 1

        for nei in adj[node]:
            print(nei) 
            print(adj[node])
            if vis[nei] == 1 and nei != parent:
                return True

            if vis[nei] == 0:
                if dfs(adj, nei, node, vis):
                    return True

        return False
```

```python
def isCycle(V, edges):
    adj = [[] for _ in range(V)]
    for src, dest in edges:
        adj[src].append(dest)   # ✅ dest
        adj[dest].append(src)   # ✅ dest
    
    vis = [0] * V
    print(adj)
    print(vis)
    for i in range(V):
        if vis[i] == 0:
            if dfs(adj, i, -1, vis):
                return True
    
    return False
```

```python
V = 4
edges = [[0, 1], [0, 2], [1, 2], [2, 3]]
isCycle(V,edges)
```
