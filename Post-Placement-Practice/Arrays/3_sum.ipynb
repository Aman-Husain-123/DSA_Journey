```python
nums = [-1,0,1,2,-1,-4]
nums.sort()

ans = []

for i in range(len(nums) - 2):
    p1 = i + 1
    p3 = len(nums) - 1

    while p1 < p3:
        curr_sum = nums[i] + nums[p1] + nums[p3]

        if curr_sum > 0:
            p3 -= 1
        elif curr_sum < 0:
            p1 += 1
        else:
            ans.append([nums[i], nums[p1], nums[p3]])
            p1 += 1
            p3 -= 1

print(ans)
```
