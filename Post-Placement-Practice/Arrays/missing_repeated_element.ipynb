```python
grid = [[1,3],[2,2]]
ans = []
s = set()

n = len(grid)
exp_sum = 0
actual_sum = 0

for i in range(n):                  # For rows
    for j in range(n):              # For cols
        actual_sum += grid[i][j]

        if grid[i][j] in s:
            a = grid[i][j]
            ans.append(a)
        s.add(grid[i][j])

exp_sum = (n * n) * ((n * n) + 1) // 2
b = exp_sum + a - actual_sum
ans.append(b)
print(ans)
```
