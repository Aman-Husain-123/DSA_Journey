```python
# Brute Approach
arr = [32,321,4,320,12]
arr.sort()

largest = float('-inf')
second = float('-inf')

for i in range(len(arr)-2,-1,-1):
    if arr[i] != largest:
        second = arr[i]
        break

print(second)


# After Sorting :- [4,12,32,320,321]
# Largest element :- 321
# if second last and last element is same then go for reverse traversal otherwise return 2nd largest
```

```python
# TODO: Better Approach
# TC :- for 1st pass -->  O(n) + for 2nd pass -->  O(n) = O(2n)
arr = [32,321,4,320,12]
largest = float('-inf')
second = float('-inf')


# 1st Pass
for i in range(len(arr)):
    if arr[i] > largest:
        largest = arr[i]


# 2nd Pass
for j in range(len(arr)):
    if arr[j] != largest and arr[j] > second:
        second = arr[j]

print(second)
```

```python
# Optimal approach
class Solution:
    def getSecondLargest(self, arr):
        # code here
        largest = arr[0]
        slargest = float('-inf')
        
        for i in range(1,len(arr)):
            if arr[i] > largest:
                slargest = largest
                largest = arr[i]
            elif arr[i] < largest and arr[i] > slargest:
                slargest = arr[i]
        
        return slargest

arr = [12, 35, 1, 10, 34, 1]
obj = Solution()
obj.getSecondLargest(arr)
```
