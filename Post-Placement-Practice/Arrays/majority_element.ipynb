```python
nums = [6,5,5]
n = len(nums) // 2
mapp = {}

for i in nums:
    mapp[i] = mapp.get(i,0) + 1 

print(mapp)
for key,value in mapp.items():
    if value > n:
        print(key)
        break
```

```python
# Better approach
nums = [6,5,5]
n = len(nums) // 2

for i in range(len(nums)):
    count = 0
    for j in range(len(nums)):
        if nums[i] == nums[j]:
            count += 1
    
    if count > n:
        print(nums[i])
        break
```

```python
# Optimal Approach
nums = [6,5,5]
ans = 0
freq = 0

for i in range(len(nums)):
    if freq == 0:
        ans = nums[i]
    
    if ans == nums[i]:
        freq += 1
    else:
        freq -= 1

count = 0
for i in nums:
    if i == ans:
        count += 1

if count > len(nums) // 2:
    print(ans)
else:
    print(-1)
# ans
```
