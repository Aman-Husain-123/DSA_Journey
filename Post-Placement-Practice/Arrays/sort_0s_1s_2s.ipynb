```python
# TODO: Better Approach
nums = [2,0,2,1,1,0]
n = len(nums)
for i in range(n):
    for j in range(i + 1,n):
        if nums[i] > nums[j]:
            temp = nums[i]
            nums[i] = nums[j]
            nums[j] = temp
print(nums)
```
