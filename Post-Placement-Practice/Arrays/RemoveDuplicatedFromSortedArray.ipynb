```python
from typing import List
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        unique = set()

        for i in range(len(nums)):
            unique.add(nums[i])

        unique = list(unique)
        unique.sort()

        for i in range(len(unique)):
            nums[i] = unique[i]

        return len(unique)

nums = [1,1,2]
obj = Solution()
obj.removeDuplicates(nums)
```
